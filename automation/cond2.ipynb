```python
import mne
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd
from collections import Counter
import warnings

# 파일 경로 설정
file_path = r"1018 승욱 시각.bdf" # 또는 .bdf 파일

# 파일 확장자에 따라 적절한 함수 사용
file_extension = os.path.splitext(file_path)[1].lower()

# --- 파일 로드 ---
try:
    if file_extension == '.bdf':
        raw = mne.io.read_raw_bdf(file_path, preload=True, verbose=False)  # stim_channel 지정 안 함
        print("BDF 파일이 로드되었습니다.")
    elif file_extension == '.set':
        raw = mne.io.read_raw_eeglab(file_path, preload=True, verbose=False)
        print("SET 파일이 로드되었습니다.")
    else:
        raise ValueError(f"지원되지 않는 파일 형식입니다: {file_extension}")
except FileNotFoundError:
    print(f"오류: 파일을 찾을 수 없습니다: {file_path}")
    # 더미 데이터 생성 또는 프로그램 종료 (여기서는 종료)
    exit()

# ==========================================================
# ✅ [1.5] T3/T4 채널 이름 강제 수정 (위치 오류 해결)
# 10-20 시스템의 T3/T4는 현대 10-10/10-5 시스템에서 T7/T8로 불립니다.
# 몽타주 좌표 인식을 위해 이름을 강제로 수정합니다.
# ==========================================================
mapping = {}
if 'T3' in raw.ch_names:
    raw.rename_channels({'T3': 'T7'})
    mapping['T3'] = 'T7'
if 'T4' in raw.ch_names:
    raw.rename_channels({'T4': 'T8'})
    mapping['T4'] = 'T8'

if mapping:
    print(f"채널 이름 수정: {mapping} (몽타주 위치 정확도 향상)")
# ==========================================================


print("모든 채널 목록:", raw.ch_names)
print(f"샘플링 주파수: {raw.info['sfreq']} Hz, 총 길이: {raw.n_times/raw.info['sfreq']:.1f} s")

# --- 채널 타입 정리(옵션): EOG로 지정하면 나중에 picks로 쉽게 제외 가능 ---
type_map = {}
for cand in ['ExG 1','ExG 2','EOG','EOG1','EOG2']:
    if cand in raw.ch_names:
        type_map[cand] = 'eog'
if type_map:
    # --- 수정: verbose 인자 제거 ---
    raw.set_channel_types(type_map)

# --- 불필요 채널 제거 ---
channels_to_remove = ['ACC22','ACC23','ACC24','Packet Counter']  # 가속도나 패킷 카운터 등
existing_channels_to_remove = [ch for ch in channels_to_remove if ch in raw.ch_names]
if existing_channels_to_remove:
    # --- 수정: verbose 인자 제거 ---
    raw.drop_channels(existing_channels_to_remove)
    print(f"제거된 채널: {existing_channels_to_remove}")


# --- 몽타주: cap에 맞게 선택 및 채널 위치 시각화 ---
montage_set_successfully = False
try:
    # BioSemi 특유 이름이 있는지 확인
    is_biosemi = any(name.startswith('A') or name.startswith('B') for name in raw.ch_names) and len(raw.ch_names) > 32

    if is_biosemi:
        montage = mne.channels.make_standard_montage('biosemi64')
        print("BioSemi 64-channel 몽타주를 시도합니다.")
    else:
        # 32채널 이하이거나 표준 명명법일 경우
        montage = mne.channels.make_standard_montage('standard_1020')
        print("Standard 10-20 몽타주를 시도합니다.")
    
    # 몽타주 적용
    # --- 수정: verbose 인자 제거 ---
    raw.set_montage(montage, on_missing='warn')
    print("몽타주가 성공적으로 설정되었습니다.")
    montage_set_successfully = True

except Exception as e:
    print(f"몽타주 설정 경고/오류: {e}. 채널 위치 시각화가 어려울 수 있습니다.")


# ==========================================================
# ✅ 채널 위치 시각화 (추가된 부분)
# ==========================================================
if montage_set_successfully:
    try:
        # 2D 평면 위에 채널 위치(topomap)를 플롯
        raw.plot_sensors(
            kind='topomap',
            title='EEG Channel Locations (Topomap) - All Channels',
            show_names=True,
            sphere='auto'
        )
        plt.show()
        print("\n[채널 위치 시각화] Topomap 플롯이 생성되었습니다.")
    except Exception as e:
        print(f"\n[경고] 채널 위치 시각화 실패: {e}")
        print("몽타주는 설정되었으나, 채널 좌표가 누락되었거나 플롯에 오류가 있을 수 있습니다.")
else:
    print("\n[알림] 몽타주 설정 실패로 채널 위치 시각화는 건너뜁니다.")
# ==========================================================


# --- 이벤트 추출 ---
def extract_events_safe(raw, file_ext):
    # 1) EEGLAB: annotations 우선
    if file_ext == '.set':
        if len(raw.annotations) > 0:
            ev, ev_id = mne.events_from_annotations(raw)
            print(f"SET/Annotations 기반 이벤트: {len(ev)}개, event_id={ev_id}")
            return ev
        else:
            print("경고: SET에 annotations가 없습니다. stim 채널 탐색을 시도합니다.")

    # 2) BDF 혹은 stim 채널이 있는 경우: 자극 채널 이름 자동 탐색
    stim_name = None
    for cand in ('Status', 'STI 014', 'TRIGGER', 'Stim', 'STIM'):
        if cand in raw.ch_names:
            stim_name = cand
            break
    if stim_name is None:
        stim_picks = mne.pick_types(raw.info, stim=True)
        if len(stim_picks) > 0:
            stim_name = raw.ch_names[stim_picks[0]]

    if stim_name is None:
        raise RuntimeError("자극(stim) 채널을 찾을 수 없습니다. (.set이면 annotations 확인 필요)")

    ev = mne.find_events(
        raw,
        stim_channel=stim_name,
        shortest_event=1,
        consecutive='increasing',
        mask=0xFF,      # 하위 8비트만 사용 (BioSemi 권장 패턴)
        mask_type='and',
        verbose=False
    )
    print(f"STIM 채널('{stim_name}') 기반 이벤트: {len(ev)}개")
    return ev

try:
    events = extract_events_safe(raw, file_extension)
except Exception as e:
    print(f"이벤트 추출 실패: {e}")
    events = np.empty((0,3), dtype=int)

# --- 이벤트 요약 출력 및 블록 길이 계산 ---
if len(events) > 0:
    codes = events[:,2]
    print(f"\n추출된 이벤트 수: {len(events)}")
    print("코드 분포:", dict(Counter(codes)))
    
    print("\n--- 이벤트 상세 정보 (시점 및 코드) ---")
    sfreq = raw.info['sfreq']
    for event in events:
        sample = event[0]
        event_code = event[2]
        timestamp_sec = sample / sfreq
        print(f"시점: {timestamp_sec:.2f}s | 샘플: {sample} | 코드: {event_code}")

    # 1(rest) 이벤트의 시작점 찾기
    rest_idx = np.where(events[:, 2] == 1)[0]
    
    print("\n--- Task 블록 길이 (Rest-to-Rest) ---")
    for i in range(len(rest_idx) - 1):
        # 현재 rest 다음 이벤트(task 시작)부터 다음 rest 이벤트(task 끝)까지
        start_sample = events[rest_idx[i] + 1][0]
        end_sample = events[rest_idx[i+1]][0]
        duration_sec = (end_sample - start_sample) / sfreq
        block_label = events[rest_idx[i] + 1][2] if rest_idx[i] + 1 < len(events) else "N/A"
        print(f"블록 {i+1} (코드 {block_label}): {duration_sec:.2f}초")
    
    print("\n--- Task to Task 블록 길이 (Task_Start-to-Next_Task_Start) ---")
    for i in range(len(rest_idx) - 1):
        try:
            # 현재 rest 다음 이벤트(task 시작)
            start_sample = events[rest_idx[i] + 1][0]
            # 다음 rest 다음 이벤트(다음 task 시작)
            end_sample = events[rest_idx[i+1] + 1][0]
            duration_sec = (end_sample - start_sample) / sfreq
            
            # 'Task-to-Task' 블록의 첫 번째 이벤트 라벨
            block_label = events[rest_idx[i] + 1][2]
            print(f"Task to Task 블록 {i+1} (코드 {block_label}): {duration_sec:.2f}초")
        except IndexError:
            # 마지막 rest 이후에 task가 없어 IndexError 발생 시
            print(f"경고: Task to Task 블록 {i+1}의 끝점을 찾을 수 없습니다. (마지막 블록)")
            continue
    
    # (옵션) 이벤트 시각화
    mne.viz.plot_events(events, sfreq=raw.info['sfreq'])
    # plt.show() # 아래 코드에서 최종적으로 한 번만 호출하기 위해 주석 처리
else:
    print("\n추출된 이벤트가 없습니다.")


# ==========================================================
# ✅ [2] 선택 채널 전처리 및 시각화 (12 Channels) - 기존 코드
# T3->T7, T4->T8로 이름이 수정되었으므로, 선택 채널 리스트도 T7, T8을 사용합니다.
# ==========================================================
print("\n" + "="*50)
print("--- [2] 선택 채널 전처리 (12 Channels) ---")

# F3/F4/F7/F8/Fp1/Fp2/Fz/C3/C4/Cz/T7/T8 (12개)
selected_channels_12 = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2']
raw_channels = raw.ch_names # 채널 이름이 T7, T8로 수정된 상태

# 실제 데이터에 존재하는 채널만 필터링 (오류 방지)
existing_selected_channels_12 = [ch for ch in selected_channels_12 if ch in raw_channels]

if not existing_selected_channels_12:
    print(f"\n경고: 요청하신 12개 채널 중 데이터에 존재하는 채널이 없습니다. 전처리를 건너뜁니다.")
    print(f"요청 채널: {selected_channels_12}")
else:
    if len(existing_selected_channels_12) < len(selected_channels_12):
        print(f"\n경고: 요청하신 12개 채널 중 {len(selected_channels_12) - len(existing_selected_channels_12)}개가 데이터에 없어 제외되었습니다.")
        print(f"실제 전처리 채널: {existing_selected_channels_12}")

    # 선택한 채널만 복사해서 새 객체로 만들기 
    raw_selected = raw.copy().pick_channels(existing_selected_channels_12)
    
    # 몽타주 시각화
    if montage_set_successfully:
        try:
            # 선택된 채널의 위치 시각화
            raw_selected.plot_sensors(
                show_names=True, #채널의 이름을 그림에 표시
                kind='topomap', 
                sphere='auto', # MNE가 자동으로 헤드 원 추정
                title='Selected 12 Channel Locations (T7/T8 Renamed)'
            )
        except Exception as e:
            print(f"\n[경고] 12개 채널 위치 시각화 실패: {e}")
    else:
        print("\n[알림] 원본 raw에 몽타주가 설정되지 않아 12개 채널 위치 시각화를 건너뜁니다.")


    # 필터 제거 전 PSD 시각화
    plt.figure(figsize=(10, 4)) # 그래프 크기 지정
    raw_selected.plot_psd(fmax=100, show=False) # 0~100Hz 범위의 PSD를 그림
    plt.title('PSD Before FIR Filtering (12 Selected Channels)') # 그래프 제목
    
    # FIR 대역 통과필터: 0.5Hz~ 55Hz
    print("\nFIR 필터링 (0.5Hz ~ 55Hz) 시작...")
    raw_selected.filter(
        l_freq=0.5,      # 저주파 차단 주파수
        h_freq=55.0,     # 고주파 차단 주파수
        method='fir',    # FIR 필터 사용
        fir_design='firwin', # FIR 설계 방식
        phase='zero',    # 위상 지연 보정 (양방향 필터링)
    )
    print("FIR 필터링 완료.")

    # 필터 제거 후 PSD 시각화
    plt.figure(figsize=(10, 4))
    raw_selected.plot_psd(fmax=100, show=False)
    plt.title('PSD After FIR Filtering (12 Selected Channels)')

# ==========================================================
# ✅ [3] 선택 채널 전처리 및 시각화 (19 Channels) - 수정된 부분
# ==========================================================
print("\n" + "="*50)
print("--- [3] 선택 채널 전처리 (19 Channels) ---")

# 10-20 국제 표준 19개 채널 (사용자 요청 목록)
# Fz, F3, F4, F7, F8, Cz, C3, C4, T7, T8, Fp1, Fp2, Pz, P3, P4, P7, P8, O1, O2
selected_channels_19 = [
    'Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 
    'Fp1', 'Fp2', 'Pz', 'P3', 'P4', 'P7', 'P8', 'O1', 'O2'
]

raw_channels = raw.ch_names # 채널 이름이 T7, T8로 수정된 상태

# 실제 데이터에 존재하는 채널만 필터링 (오류 방지)
existing_selected_channels_19 = [ch for ch in selected_channels_19 if ch in raw_channels]

# raw19 변수에 19개 채널 데이터 복사 및 선택
if not existing_selected_channels_19:
    print(f"\n경고: 요청하신 19개 채널 중 데이터에 존재하는 채널이 없습니다. 전처리를 건너뜁니다.")
    print(f"요청 채널: {selected_channels_19}")
    # 데이터가 없으므로 raw19를 생성하지 않습니다.
else:
    if len(existing_selected_channels_19) < len(selected_channels_19):
        print(f"\n경고: 요청하신 19개 채널 중 {len(selected_channels_19) - len(existing_selected_channels_19)}개가 데이터에 없어 제외되었습니다.")
        print(f"실제 전처리 채널: {existing_selected_channels_19}")

    # 선택한 채널만 복사해서 새 객체 raw19로 만들기 
    raw19 = raw.copy().pick_channels(existing_selected_channels_19)
    print(f"raw19 객체 생성 완료. 포함 채널: {raw19.ch_names}")
    
    # 몽타주 시각화
    if montage_set_successfully:
        try:
            # 선택된 채널의 위치 시각화
            raw19.plot_sensors(
                show_names=True,
                kind='topomap', 
                sphere='auto',
                title='Selected 19 Channel Locations (User Defined)'
            )
        except Exception as e:
            print(f"\n[경고] 19개 채널 위치 시각화 실패: {e}")
    else:
        print("\n[알림] 원본 raw에 몽타주가 설정되지 않아 19개 채널 위치 시각화를 건너뜁니다.")

    # 필터 제거 전 PSD 시각화
    plt.figure(figsize=(10, 4))
    raw19.plot_psd(fmax=100, show=False)
    plt.title('PSD Before FIR Filtering (19 Selected Channels)')
    
    # FIR 대역 통과필터: 0.5Hz~ 55Hz
    print("\nraw19 FIR 필터링 (0.5Hz ~ 55Hz) 시작...")
    raw19.filter(
        l_freq=0.5,
        h_freq=55.0,
        method='fir',
        fir_design='firwin',
        phase='zero',
    )
    print("raw19 FIR 필터링 완료.")

    # 필터 제거 후 PSD 시각화
    plt.figure(figsize=(10, 4))
    raw19.plot_psd(fmax=100, show=False)
    plt.title('PSD After FIR Filtering (19 Selected Channels)')

plt.show() # 모든 시각화 결과(이벤트, 몽타주, PSD 전후)를 한 번에 plot
# ==========================================================
```

```python
import numpy as np
import mne

# =========================================================================
# ⚠️ 주의: 아래 코드에서는 'raw', 'raw_avg_ref', 'raw_avg_ref19' 변수가 이미 정의되어 있다고 가정합니다.
# (예: raw = mne.io.read_raw_bdf('your_file.bdf', preload=True))
# =========================================================================

# -----------------
# 1. 이벤트 추출 (raw 객체 사용) - X와 X19 모두 이 이벤트를 사용
# -----------------
# raw 객체에서 이벤트 정보를 추출합니다.
# BDF 파일은 Annotation이 아닌 Status 채널에 이벤트를 저장하는 경우가 많습니다.

# 이벤트 추출을 위해 사용할 raw 객체
raw_for_events = raw

# sfreq는 raw_avg_ref의 샘플링 주파수
# (raw_avg_ref19도 동일한 sfreq를 가질 것으로 가정)
try:
    sfreq = raw_avg_ref.info['sfreq']
    print(f"[OK] raw_avg_ref 변수가 확인되었습니다. | sfreq={sfreq} Hz")
except NameError:
    try:
        sfreq = raw_for_events.info['sfreq']
        print(f"[OK] raw_for_events 변수에서 sfreq={sfreq} Hz 확인.")
    except Exception:
        print("[ERROR] sfreq를 확인할 수 없습니다. raw_avg_ref 또는 raw 변수를 확인하세요.")
        exit()

# 1) 리샘플된 raw에서 이벤트 다시 추출 (annotations 우선)
if len(raw_for_events.annotations) > 0:
    # 1-1. Annotation이 있는 경우
    events, _ = mne.events_from_annotations(raw_for_events)
else:
    # 1-2. BDF의 Status/Trigger 채널에서 이벤트 찾기
    stim_channel_names = ('Status','STI 014','TRIGGER','Stim','STIM')
    stim_name = next((c for c in stim_channel_names
                      if c in raw_for_events.ch_names), None)

    # BDF의 경우, Biosemi는 16비트 채널을 사용하지만 하위 8비트에만 이벤트가 기록되는 경우가 흔하므로
    # 'mask=0xFF' (255)를 사용하여 하위 8비트만 마스킹합니다.
    events = (mne.find_events(raw_for_events, stim_channel=stim_name,
                              shortest_event=1, consecutive='increasing',
                              mask=0xFF, mask_type='and', verbose=False)
              if stim_name else np.empty((0,3), int))

# --- 중간 과정 변수 확인 (events) ---
print("-" * 40)
print("  중간 과정 변수 확인")
print("-" * 40)
print(f"▶ events (추출된 이벤트):")
print(f"  - 차원(shape): {events.shape}")
if events.size > 0:
    print(f"  - 첫 5개 이벤트: \n{events[:5]}\n")
else:
    print("  - 이벤트가 감지되지 않았습니다. 코드 실행을 멈춥니다.\n")
    exit() # 이벤트 없으면 이후 과정 무의미

# -------------------------------------


# -----------------
# 2. 트라이얼 정의 및 에포크 슬라이싱 기준 설정
# -----------------
# 2) 부정(2)과 긍정(3) 이벤트만 필터링하여 trial 정의
NEG, POS = 2, 3
trials = events[np.isin(events[:, 2], [NEG, POS])]

# --- trials 변수 확인 ---
print(f"▶ trials (정의된 부정/긍정 트라이얼):")
print(f"  - 총 트라이얼 수: {len(trials)}")
if len(trials) == 0:
    print("  - 유효한 트라이얼이 정의되지 않았습니다. 코드 실행을 멈춥니다.\n")
    exit()
else:
    print(f"  - 첫 5개 트라이얼: \n{trials[:5]}\n")
print("-" * 40)

# 트라이얼/에포크 공통 설정
tmin = -2.0
tmax = 9.0
EPOCH_LEN = 2.0
EPOCH_STEP = 1.0
win = int(round(EPOCH_LEN * sfreq))
step = int(round(EPOCH_STEP * sfreq))
BASELINE_RANGE = (-2.0, 0.0)
CROP_RANGE = (1.0, 9.0)


# ======================================================================
# ✅ [A] raw_avg_ref (12채널) 데이터 처리
# ======================================================================
print("\n" + "="*50)
print("--- [A] raw_avg_ref (12채널) 데이터 처리 ---")

# (12채널) MNE 객체 생성 및 슬라이딩 윈도우 분할 -> X, y, groups 생성
pick = mne.pick_types(raw_avg_ref.info, eeg=True, exclude='bads')

epochs_full = mne.Epochs(
    raw_avg_ref,
    trials,
    event_id={ 'NEG': NEG, 'POS': POS },
    tmin=tmin,
    tmax=tmax,
    baseline=None,
    picks=pick,
    preload=True,
    verbose=False
)

print(f"[MNE] epochs_full 객체에 apply_baseline 적용 중: {BASELINE_RANGE}s")

# 분석 구간(1.0s ~ 9.0s)으로 Crop
epochs_cropped = epochs_full.copy().crop(tmin=CROP_RANGE[0], tmax=CROP_RANGE[1])

# NumPy 배열로 추출 (X_trials: N_trials x N_channels x 8s_samples)
X_trials = epochs_cropped.get_data()
y_trials = epochs_cropped.events[:, 2]
groups_trials = np.asarray([trial[0] for trial in trials], dtype=np.int64)

# 3) 각 트라이얼 내를 고정 길이 에폭으로 슬라이싱 (2-9초)
X_list, y_list, groups_list = [], [], []
max_idx = X_trials.shape[2]

for i in range(X_trials.shape[0]):
    lab = 0 if y_trials[i] == NEG else 1
    trial_start_samp = groups_trials[i]

    s = 0 # 이미 1.0s에서 Crop된 데이터이므로 0부터 시작

    while s + win <= max_idx:
        seg = X_trials[i, :, s : s + win]
        X_list.append(seg)
        y_list.append(lab)
        groups_list.append(trial_start_samp)
        s += step

X = np.stack(X_list, axis=0) if X_list else np.empty((0, len(pick), win), np.float32)
y = np.asarray(y_list, dtype=np.int64)
groups = np.asarray(groups_list, dtype=np.int64)

# --- 최종 변수 상세 확인 (X) ---
print(f"\n[결과] X (12채널 에포크 데이터셋) 준비 완료")
print(f"  - X.shape: {X.shape}")
print(f"  - y.shape: {y.shape}")
print(f"  - groups.shape: {groups.shape}")


# ======================================================================
# ✅ [B] raw_avg_ref19 (19채널) 데이터 처리 - y19, groups19 생성 부분 제거됨
# ======================================================================
print("\n" + "="*50)
print("--- [B] raw_avg_ref19 (19채널) 데이터 처리 ---")

try:
    pick19 = mne.pick_types(raw_avg_ref19.info, eeg=True, exclude='bads')
except NameError:
    print("[ERROR] raw_avg_ref19 변수를 찾을 수 없습니다. 19채널 처리 과정을 건너뜁니다.")
    # X19를 빈 배열로 정의 (y19, groups19는 생성하지 않음)
    X19 = np.empty((0, 19, win), np.float32)
else:
    epochs_full19 = mne.Epochs(
        raw_avg_ref19,
        trials, # <-- 동일한 이벤트 사용
        event_id={ 'NEG': NEG, 'POS': POS },
        tmin=tmin,
        tmax=tmax,
        baseline=None,
        picks=pick19,
        preload=True,
        verbose=False
    )

    print(f"[MNE] epochs_full19 객체에 apply_baseline 적용 중: {BASELINE_RANGE}s")

    # 분석 구간(1.0s ~ 9.0s)으로 Crop
    epochs_cropped19 = epochs_full19.copy().crop(tmin=CROP_RANGE[0], tmax=CROP_RANGE[1])

    # NumPy 배열로 추출 (X_trials19: N_trials x N_channels x 8s_samples)
    X_trials19 = epochs_cropped19.get_data()
    y_trials19 = epochs_cropped19.events[:, 2] # 원본 이벤트 레이블
    groups_trials19 = np.asarray([trial[0] for trial in trials], dtype=np.int64) # 원본 이벤트 샘플

    # 3) 각 트라이얼 내를 고정 길이 에폭으로 슬라이싱 (2-9초)
    X_list19 = [] # y_list19, groups_list19는 y_list, groups_list와 동일하므로 제거
    max_idx19 = X_trials19.shape[2]

    # y와 groups가 12채널 데이터 처리 과정에서 이미 생성되었으므로,
    # 19채널에서는 X19만 생성하고, 레이블/그룹은 y/groups를 공유합니다.
    # 단, y와 groups를 업데이트하지 않기 위해 X_list19만 채웁니다.
    
    # ⚠️ 경고: 이 방식은 12채널과 19채널의 에포크 수가 정확히 같을 때만 유효합니다.
    # 채널 개수가 다를 경우 에포크 생성에 실패할 가능성은 낮으므로, 동일하다고 가정합니다.

    for i in range(X_trials19.shape[0]):
        # lab19, trial_start_samp19 변수는 더 이상 y_list19, groups_list19에 사용되지 않습니다.
        
        s = 0 # 이미 1.0s에서 Crop된 데이터이므로 0부터 시작

        while s + win <= max_idx19:
            seg19 = X_trials19[i, :, s : s + win]
            X_list19.append(seg19)
            # y_list19.append(lab19) <-- 제거됨
            # groups_list19.append(trial_start_samp19) <-- 제거됨
            s += step

    X19 = np.stack(X_list19, axis=0) if X_list19 else np.empty((0, len(pick19), win), np.float32)
    # y19 = np.asarray(y_list19, dtype=np.int64) <-- 제거됨
    # groups19 = np.asarray(groups_list19, dtype=np.int64) <-- 제거됨

    # --- 최종 변수 상세 확인 (X19) ---
    print(f"\n[결과] X19 (19채널 에포크 데이터셋) 준비 완료")
    print(f"  - X19.shape: {X19.shape}")
    print(f"  - y (공유 레이블) shape: {y.shape}") # y와 groups 공유 표시
    print(f"  - groups (공유 그룹) shape: {groups.shape}")
    if X19.size > 0 and y.size > 0:
        unique_y, counts_y = np.unique(y, return_counts=True)
        print(f"  - y 레이블 분포: {dict(zip(unique_y, counts_y))}")
# ======================================================================

# --- 최종 변수 정리 (X와 X19 결과 비교) ---
print("\n" + "="*50)
print("--- 최종 결과 비교 ---")
print(f"X (12채널 데이터셋) shape: {X.shape}")
print(f"X19 (19채널 데이터셋) shape: {X19.shape}")
print(f"y (레이블) shape: {y.shape}")
print(f"groups (그룹 ID) shape: {groups.shape}")
print("="*50)
```

```python
import numpy as np
from scipy.signal import welch

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 MNE 데이터 준비 코드에서 이미 정의되어야 합니다.
# epochs_full: MNE Epochs 객체 (12채널, tmin=-2.0s 포함)
# epochs_full19: MNE Epochs 객체 (19채널, tmin=-2.0s 포함) <--- 19채널용 변수 추가
# sfreq: 샘플링 주파수
# N_EPOCHS_PER_TRIAL: 트라이얼 당 에포크 수 (예: 7)
# =========================================================================

# -------------------------------------------------------------------------
# 1. 주파수 밴드 설정
# -------------------------------------------------------------------------
BANDS = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
N_EPOCHS_PER_TRIAL = 7 # 이 변수는 외부에서 정의되었다고 가정
eps_stability = 1e-12

# -------------------------------------------------------------------------
# 2. 베이스라인 Log-PSD 및 DE 특징 계산 함수 정의
# -------------------------------------------------------------------------
def compute_baseline_features(X_base, sfreq, bands):
    """
    각 트라이얼 및 채널에 대해 주파수 밴드별 Log-PSD 특징(Mean, Median, Peak)
    및 DE 특징을 계산합니다.
    
    반환: F_mean_base, F_median_base, F_peak_base, F_de_base
          (4개의 (N_trials, N_channels, N_bands) 배열)
    """
    n_trial, n_ch, n_samp_base = X_base.shape
    n_b = len(bands)
    
    # Welch 윈도우 크기 설정 (1초)
    nperseg = int(sfreq * 1.0)
    
    # 결과 배열 초기화: 4가지 특징
    F_mean_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32)
    F_median_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32)
    F_peak_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32)
    F_de_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32) # 🌟 DE 배열 추가

    for i in range(n_trial):
        for ch in range(n_ch):
            channel_data = X_base[i, ch, :]
            
            # Welch PSD 계산
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12  # V^2를 µV^2로 변환
            
            # 주파수 해상도 (DE 계산에 필요)
            df = f[1] - f[0] if len(f) > 1 else 1.0
            
            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                idx = np.logical_and(f >= low, f <= high)
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    # 1. Log-PSD 기반 특징 (Mean, Median, Peak)
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_base[i, ch, b_idx] = np.mean(band_log_psd)
                    F_median_base[i, ch, b_idx] = np.median(band_log_psd)
                    F_peak_base[i, ch, b_idx] = np.max(band_log_psd)
                    
                    # 2. 🌟 밴드 파워 및 DE 기반 특징 (DE) 추가
                    band_power = np.sum(band_psd) * df
                    
                    if band_power > 0:
                        F_de_base[i, ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability)
                    else:
                        F_de_base[i, ch, b_idx] = 0.0
                    
                else:
                    # 데이터가 없는 경우 0으로 처리
                    F_mean_base[i, ch, b_idx] = 0.0
                    F_median_base[i, ch, b_idx] = 0.0
                    F_peak_base[i, ch, b_idx] = 0.0
                    F_de_base[i, ch, b_idx] = 0.0 # 🌟 DE 초기화
                    
    # 🌟 4개 변수 반환
    return F_mean_base, F_median_base, F_peak_base, F_de_base

# -------------------------------------------------------------------------
# 3. 베이스라인 데이터 추출 및 특징 계산 (12채널)
# -------------------------------------------------------------------------
print("="*50)
print("--- [A] 12채널 베이스라인 특징 계산 ---")
BASELINE_RANGE = (-2.0, 0.0)

# 12채널 베이스라인 데이터 추출
try:
    epochs_baseline = epochs_full.copy().crop(tmin=BASELINE_RANGE[0], tmax=BASELINE_RANGE[1])
    X_baseline = epochs_baseline.get_data() # (N_trials, 12_ch, N_samp_baseline)
except NameError:
    print("[경고] epochs_full 변수가 정의되지 않아 12채널 처리를 건너뜁니다.")
    X_baseline = np.empty((0,12,int(sfreq*2)))
except Exception as e:
    print(f"[오류] 12채널 베이스라인 추출 중 오류 발생: {e}")
    X_baseline = np.empty((0,12,int(sfreq*2)))


# 🌟 12채널 특징 계산 실행
print("[OK] 12채널 베이스라인 Log-PSD 및 DE 특징 계산 중...")
F_mean_baseline_trials, F_median_baseline_trials, F_peak_baseline_trials, F_de_baseline_trials = \
    compute_baseline_features(X_baseline, sfreq, BANDS)

print("[OK] 12채널 베이스라인 Log-PSD 및 DE 특징 계산 완료")
print(f"▶ F_mean_baseline_trials (트라이얼 평균): {F_mean_baseline_trials.shape}")
# ... (다른 트라이얼 변수 출력 생략)

# 🌟 12채널 특징 에포크 매칭
F_mean_baseline_matched = np.repeat(F_mean_baseline_trials, N_EPOCHS_PER_TRIAL, axis=0)
F_median_baseline_matched = np.repeat(F_median_baseline_trials, N_EPOCHS_PER_TRIAL, axis=0)
F_peak_baseline_matched = np.repeat(F_peak_baseline_trials, N_EPOCHS_PER_TRIAL, axis=0)
F_de_baseline_matched = np.repeat(F_de_baseline_trials, N_EPOCHS_PER_TRIAL, axis=0)

print(f"▶ F_mean_baseline_matched (매칭된 평균): {F_mean_baseline_matched.shape}")
# ... (다른 매칭 변수 출력 생략)


# -------------------------------------------------------------------------
# 4. 베이스라인 데이터 추출 및 특징 계산 (19채널) - 추가된 부분
# -------------------------------------------------------------------------
print("="*50)
print("--- [B] 19채널 베이스라인 특징 계산 ---")

# 19채널 베이스라인 데이터 추출
try:
    epochs_baseline19 = epochs_full19.copy().crop(tmin=BASELINE_RANGE[0], tmax=BASELINE_RANGE[1])
    X_baseline19 = epochs_baseline19.get_data() # (N_trials, 19_ch, N_samp_baseline)
except NameError:
    print("[경고] epochs_full19 변수가 정의되지 않아 19채널 처리를 건너뜁니다.")
    X_baseline19 = np.empty((0,19,int(sfreq*2)))
except Exception as e:
    print(f"[오류] 19채널 베이스라인 추출 중 오류 발생: {e}")
    X_baseline19 = np.empty((0,19,int(sfreq*2)))


# 🌟 19채널 특징 계산 실행
print("[OK] 19채널 베이스라인 Log-PSD 및 DE 특징 계산 중...")
F_mean_baseline_trials19, F_median_baseline_trials19, F_peak_baseline_trials19, F_de_baseline_trials19 = \
    compute_baseline_features(X_baseline19, sfreq, BANDS)

print("[OK] 19채널 베이스라인 Log-PSD 및 DE 특징 계산 완료")
print(f"▶ F_mean_baseline_trials19 (트라이얼 평균): {F_mean_baseline_trials19.shape}")
print(f"▶ F_de_baseline_trials19 (트라이얼 DE): {F_de_baseline_trials19.shape}")

# 🌟 19채널 특징 에포크 매칭
F_mean_baseline_matched19 = np.repeat(F_mean_baseline_trials19, N_EPOCHS_PER_TRIAL, axis=0)
F_median_baseline_matched19 = np.repeat(F_median_baseline_trials19, N_EPOCHS_PER_TRIAL, axis=0)
F_peak_baseline_matched19 = np.repeat(F_peak_baseline_trials19, N_EPOCHS_PER_TRIAL, axis=0)
F_de_baseline_matched19 = np.repeat(F_de_baseline_trials19, N_EPOCHS_PER_TRIAL, axis=0)

print("[OK] 19채널 베이스라인 Log-PSD 및 DE 특징 에포크 매칭 완료")
print(f"▶ F_mean_baseline_matched19 (매칭된 평균): {F_mean_baseline_matched19.shape}")
print(f"▶ F_de_baseline_matched19 (매칭된 DE): {F_de_baseline_matched19.shape}")
print("="*50)
```

```python
import numpy as np

def print_group_label_details(groups, y):
    """
    각 그룹(트라이얼)의 라벨 분포를 상세히 출력합니다.
    """
    print("\n" + "="*50)
    print("      groups & y 데이터 그룹화 상세 정보")
    print("="*50)

    if groups.shape != y.shape:
        print(f"❌ 오류: groups와 y의 차원이 다릅니다. groups: {groups.shape}, y: {y.shape}")
        return

    unique_groups = np.unique(groups)
    
    # 각 그룹의 상세 정보를 출력
    for g in unique_groups:
        # 현재 그룹에 속한 에포크들의 인덱스 추출
        group_indices = np.where(groups == g)[0]
        
        # 해당 인덱스들의 라벨 값 추출
        group_labels = y[group_indices]
        
        # 라벨이 모두 같은지 확인
        unique_labels_in_group = np.unique(group_labels)
        
        status = "✅ 일치" if len(unique_labels_in_group) == 1 else "❌ 불일치"
        
        print(f"▶ 그룹 ID: {g} | 상태: {status}")
        print(f"  - 속한 에포크 수: {len(group_indices)}")
        print(f"  - 발견된 라벨: {unique_labels_in_group}")
        
        # 라벨이 불일치하는 경우 상세 분포 출력
        if status == "❌ 불일치":
            label_counts = {label: np.sum(group_labels == label) for label in unique_labels_in_group}
            print(f"  - 라벨 분포: {label_counts}")
        print("-" * 50)

    print("\n▶ 요약:")
    print(f"  - 총 그룹(트라이얼) 수: {len(unique_groups)}")
    mismatched_groups = [g for g in unique_groups if len(np.unique(y[np.where(groups == g)[0]])) > 1]
    print(f"  - 라벨 불일치 그룹 수: {len(mismatched_groups)}")
    print("="*50 + "\n")

# --------------------------------------------------------------------
# 사용 예시:
# 이 코드를 기존 스크립트의 groups와 y 변수가 생성된 후에 추가하여 호출하세요.
# 예시:
# F_3d = compute_psd_log10_epochs_3d(X, y, groups, sfreq, BANDS)
print_group_label_details(groups, y)
```

```python
import numpy as np
from scipy.signal import welch

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 이미 정의되어야 합니다.
# X: (n_epochs, n_ch, n_samp) - 12채널 슬라이싱된 이벤트 기간 EEG 데이터
# X19: (n_epochs, 19, n_samp) - 19채널 슬라이싱된 이벤트 기간 EEG 데이터
# y: (n_epochs,)
# groups: (n_epochs,)
# sfreq: 샘플링 주파수 (Hz)
#
# [이전 베이스라인 코드에서 생성된 19채널 변수들]
# F_mean_baseline_matched19: (n_epochs, 19, n_bands)
# F_median_baseline_matched19: (n_epochs, 19, n_bands)
# F_peak_baseline_matched19: (n_epochs, 19, n_bands)
# F_de_baseline_matched19: (n_epochs, 19, n_bands)
# =========================================================================

# -------------------------------------------------------------------------
# 1. 주파수 밴드 설정 (12채널 코드와 동일)
# -------------------------------------------------------------------------
BANDS = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
band_keys = list(BANDS.keys())
eps_stability = 1e-12

# -------------------------------------------------------------------------
# 2. 이벤트 기간 특징 계산 함수 (Mean, Median, Peak, DE) - 함수는 재사용
# -------------------------------------------------------------------------
def compute_event_features_3d(X, sfreq, bands):
    """
    이벤트 기간(X)에 대해 4가지 특징(Log-PSD Mean, Median, Peak 및 DE)을 계산합니다.
    """
    n_ep, n_ch, n_samp = X.shape
    n_b = len(bands)
    
    # Welch 윈도우 크기 설정 (1초)
    nperseg = int(sfreq)
    
    # 결과 배열 초기화: 4가지 특징
    F_mean_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_median_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_peak_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_de_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)

    for i in range(n_ep):
        for ch in range(n_ch):
            channel_data = X[i, ch, :]
            
            # Welch PSD 계산
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12  # V^2를 µV^2로 변환
            
            # 주파수 해상도 (DE 계산에 필요)
            df = f[1] - f[0] if len(f) > 1 else 1.0
            
            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                idx = np.logical_and(f >= low, f <= high)
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    # 1. Log-PSD 기반 특징 (Mean, Median, Peak)
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_event[i, ch, b_idx] = np.mean(band_log_psd)
                    F_median_event[i, ch, b_idx] = np.median(band_log_psd)
                    F_peak_event[i, ch, b_idx] = np.max(band_log_psd)
                    
                    # 2. 밴드 파워 및 DE 기반 특징 (DE)
                    band_power = np.sum(band_psd) * df
                    
                    if band_power > 0:
                        F_de_event[i, ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability)
                    else:
                        F_de_event[i, ch, b_idx] = 0.0
                    
                else:
                    # 데이터가 없는 경우 0으로 처리
                    F_mean_event[i, ch, b_idx] = 0.0
                    F_median_event[i, ch, b_idx] = 0.0
                    F_peak_event[i, ch, b_idx] = 0.0
                    F_de_event[i, ch, b_idx] = 0.0
                    
    return F_mean_event, F_median_event, F_peak_event, F_de_event

# -------------------------------------------------------------------------
# 3. 이벤트 기간 특징 추출 (12채널 실행)
# -------------------------------------------------------------------------
print("="*60)
print("--- [A] 12채널 특징 추출 및 정규화 ---")

print("[OK] 12채널 이벤트 기간 Log-PSD 및 DE 특징 계산 중...")
F_mean_event, F_median_event, F_peak_event, F_de_event = \
    compute_event_features_3d(X, sfreq, BANDS)
print("[OK] 12채널 이벤트 기간 특징 계산 완료.")

# Log Ratio 정규화 (12채널)
F_mean_3d = F_mean_event - F_mean_baseline_matched
F_median_3d = F_median_event - F_median_baseline_matched
F_peak_3d = F_peak_event - F_peak_baseline_matched
F_de_3d = F_de_event - F_de_baseline_matched
print("[OK] 12채널 Log Ratio 정규화 완료.")

# 비대칭 특징 계산 (12채널, F_de_3d 기반)
CHANNELS = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2']
HEM_PAIRS = [ ('F3', 'F4'), ('F7', 'F8'), ('C3', 'C4'), ('T7', 'T8'), ('Fp1', 'Fp2') ]
pair_names = [f"{L}-{R}" for L, R in HEM_PAIRS] 
ch_to_idx = {ch_name: i for i, ch_name in enumerate(CHANNELS)}
n_pairs = len(HEM_PAIRS)
n_ep, _, n_b = F_de_3d.shape
F_dasm_3d_lr = np.empty((n_ep, n_pairs, n_b), dtype=np.float32)
F_rasm_3d_lr = np.empty((n_ep, n_pairs, n_b), dtype=np.float32)

for i, (left_ch, right_ch) in enumerate(HEM_PAIRS):
    left_idx = ch_to_idx[left_ch]
    right_idx = ch_to_idx[right_ch]
    de_norm_left = F_de_3d[:, left_idx, :]
    de_norm_right = F_de_3d[:, right_idx, :]
    
    F_dasm_3d_lr[:, i, :] = de_norm_left - de_norm_right
    F_rasm_3d_lr[:, i, :] = de_norm_left / (de_norm_right + eps_stability) # 1e-9 대신 eps_stability 사용

F_dasm_3d = F_dasm_3d_lr.copy()
F_rasm_3d = F_rasm_3d_lr.copy()
print("[OK] 12채널 비대칭 특징 (DASM, RASM) 생성 완료.")
print(f"  - F_de_3d (DE 차분): {F_de_3d.shape}")
print(f"  - F_dasm_3d (DASM): {F_dasm_3d.shape}")


# -------------------------------------------------------------------------
# 4. 이벤트 기간 특징 추출 (19채널 실행) - ✨ 추가된 섹션
# -------------------------------------------------------------------------
print("="*60)
print("--- [B] 19채널 특징 추출 및 정규화 ---")

try:
    print("[OK] 19채널 이벤트 기간 Log-PSD 및 DE 특징 계산 중...")
    F_mean_event19, F_median_event19, F_peak_event19, F_de_event19 = \
        compute_event_features_3d(X19, sfreq, BANDS)
    print("[OK] 19채널 이벤트 기간 특징 계산 완료.")

    # 4-1. 로그 비율 정규화 (19채널)
    F_mean_3d19 = F_mean_event19 - F_mean_baseline_matched19
    F_median_3d19 = F_median_event19 - F_median_baseline_matched19
    F_peak_3d19 = F_peak_event19 - F_peak_baseline_matched19
    F_de_3d19 = F_de_event19 - F_de_baseline_matched19
    print("[OK] 19채널 Log Ratio 정규화 완료.")

    # 4-2. 비대칭 특징 (DASM, RASM) 계산 - 정규화된 DE(F_de_3d19) 기반
    
    # 19채널 목록 및 쌍 정의
    CHANNELS19 = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2', 'Pz', 'P3', 'P4', 'P7', 'P8', 'O1', 'O2']
    # 19채널 중에서 반구 쌍만 추출
    HEM_PAIRS19 = [ ('F3', 'F4'), ('F7', 'F8'), ('C3', 'C4'), ('T7', 'T8'), ('Fp1', 'Fp2'), ('P3', 'P4'), ('P7', 'P8'), ('O1', 'O2') ]
    pair_names19 = [f"{L}-{R}" for L, R in HEM_PAIRS19] 
    ch_to_idx19 = {ch_name: i for i, ch_name in enumerate(CHANNELS19)}
    n_pairs19 = len(HEM_PAIRS19)

    # 특징 배열 초기화
    n_ep19, _, n_b19 = F_de_3d19.shape
    F_dasm_3d_lr19 = np.empty((n_ep19, n_pairs19, n_b19), dtype=np.float32)
    F_rasm_3d_lr19 = np.empty((n_ep19, n_pairs19, n_b19), dtype=np.float32)

    # 각 채널 쌍에 대해 DASM, RASM 계산
    for i, (left_ch, right_ch) in enumerate(HEM_PAIRS19):
        # 19채널 목록 내 인덱스 사용
        left_idx = ch_to_idx19[left_ch] 
        right_idx = ch_to_idx19[right_ch]
        
        # 정규화된 DE 값 추출
        de_norm_left = F_de_3d19[:, left_idx, :]
        de_norm_right = F_de_3d19[:, right_idx, :]
        
        # DASM 계산: Left - Right
        F_dasm_3d_lr19[:, i, :] = de_norm_left - de_norm_right
        
        # RASM 계산: Left / Right
        F_rasm_3d_lr19[:, i, :] = de_norm_left / (de_norm_right + eps_stability)

    F_dasm_3d19 = F_dasm_3d_lr19.copy()
    F_rasm_3d19 = F_rasm_3d_lr19.copy()
    print("[OK] 19채널 비대칭 특징 (DASM, RASM) 생성 완료.")
    
except NameError:
    print("[경고] X19 또는 19채널 베이스라인 변수가 정의되지 않아 19채널 처리를 건너뜁니다.")
    # 변수들을 빈 배열로 정의 (오류 방지)
    F_mean_3d19 = F_median_3d19 = F_peak_3d19 = F_de_3d19 = np.empty((0, 19, len(BANDS)), np.float32)
    F_dasm_3d19 = F_rasm_3d19 = np.empty((0, 8, len(BANDS)), np.float32) # 8쌍 기준
except Exception as e:
    print(f"[오류] 19채널 특징 계산 중 오류 발생: {e}")


# -------------------------------------------------------------------------
# 5. 최종 결과 확인
# -------------------------------------------------------------------------
print("="*60)
print("--- 최종 결과 비교 ---")
print("\n[12채널 특징]")
print(f"  - F_mean_3d (Log-PSD Mean 차분): {F_mean_3d.shape}")
print(f"  - F_de_3d (DE 차분): {F_de_3d.shape}")
print(f"  - F_dasm_3d (DASM): {F_dasm_3d.shape}")
print(f"  - F_rasm_3d (RASM): {F_rasm_3d.shape}")
print(f"  - DASM/RASM 채널 쌍 순서: {pair_names}")


print("\n[19채널 특징]")
print(f"  - F_mean_3d19 (Log-PSD Mean 차분): {F_mean_3d19.shape}")
print(f"  - F_de_3d19 (DE 차분): {F_de_3d19.shape}")
print(f"  - F_dasm_3d19 (DASM): {F_dasm_3d19.shape}")
print(f"  - F_rasm_3d19 (RASM): {F_rasm_3d19.shape}")
if 'pair_names19' in locals():
    print(f"  - DASM/RASM 채널 쌍 순서: {pair_names19}")
print("="*60)
```

```python
import numpy as np
from scipy.signal import welch

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 이미 정의되어야 합니다. (이전 단계 결과 변수)
# X: (n_epochs, n_ch, n_samp) - 슬라이싱된 이벤트 기간 EEG 데이터
# y: (n_epochs,)
# groups: (n_epochs,)
# sfreq: 샘플링 주파수 (Hz)
# F_mean_baseline_matched: (n_epochs, n_ch, n_bands)
# F_median_baseline_matched: (n_epochs, n_ch, n_bands)
# F_peak_baseline_matched: (n_epochs, n_ch, n_bands)
# F_de_baseline_matched: (n_epochs, n_ch, n_bands)
# =========================================================================

# -------------------------------------------------------------------------
# 1. 주파수 밴드 설정
# -------------------------------------------------------------------------
BANDS = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
band_keys = list(BANDS.keys())
eps_stability = 1e-12
CHANNELS = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2'] # 채널 정의

# -------------------------------------------------------------------------
# 2. 이벤트 기간 특징 계산 함수 (Mean, Median, Peak, DE)
# -------------------------------------------------------------------------
def compute_event_features_3d(X, sfreq, bands):
    """
    이벤트 기간(X)에 대해 4가지 특징(Log-PSD Mean, Median, Peak 및 DE)을 계산합니다.
    (이전 코드와 동일하므로 생략)
    """
    n_ep, n_ch, n_samp = X.shape
    n_b = len(bands)
    nperseg = int(sfreq)
    
    F_mean_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_median_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_peak_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_de_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)

    for i in range(n_ep):
        for ch in range(n_ch):
            channel_data = X[i, ch, :]
            
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12
            df = f[1] - f[0] if len(f) > 1 else 1.0
            
            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                idx = np.logical_and(f >= low, f <= high)
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_event[i, ch, b_idx] = np.mean(band_log_psd)
                    F_median_event[i, ch, b_idx] = np.median(band_log_psd)
                    F_peak_event[i, ch, b_idx] = np.max(band_log_psd)
                    
                    band_power = np.sum(band_psd) * df
                    if band_power > 0:
                        F_de_event[i, ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability)
                    else:
                        F_de_event[i, ch, b_idx] = 0.0
                else:
                    F_mean_event[i, ch, b_idx] = 0.0
                    F_median_event[i, ch, b_idx] = 0.0
                    F_peak_event[i, ch, b_idx] = 0.0
                    F_de_event[i, ch, b_idx] = 0.0
                    
    return F_mean_event, F_median_event, F_peak_event, F_de_event

# -------------------------------------------------------------------------
# 3. 이벤트 기간 특징 추출 및 Log Ratio 정규화 (전제 조건의 변수 정의)
# -------------------------------------------------------------------------
# (이 부분은 전제 조건의 변수들이 정의되었다고 가정하고 실제 계산을 수행해야 합니다.)

# ⚠️ 실제 환경에서는 X, sfreq 등이 정의되어야 합니다.
# (이전 단계의 코드를 여기에 복사하여 붙여넣고 실행하는 것을 가정합니다.)
# F_mean_event, F_median_event, F_peak_event, F_de_event = compute_event_features_3d(X, sfreq, BANDS)
# F_mean_3d = F_mean_event - F_mean_baseline_matched
# F_median_3d = F_median_event - F_median_baseline_matched
# F_peak_3d = F_peak_event - F_peak_baseline_matched
# F_de_3d = F_de_event - F_de_baseline_matched
# F_dasm_3d, F_rasm_3d 계산...

# ------------------------------------------------------------------------------
# ⭐ 12개 단독 채널 특징 변수 추출 및 개별 변수 저장 로직 (수정할 필요 없음) ⭐
# ------------------------------------------------------------------------------

# 4가지 특징 유형 정의
FEATURE_TYPES = ['mean', 'median', 'peak', 'de']
# 채널 이름을 인덱스로 변환하는 딕셔너리
ch_to_idx = {ch_name: i for i, ch_name in enumerate(CHANNELS)}

# 3D 특징 배열 매핑 (Log Ratio 정규화된 특징 변수 사용)
feat_arrays = {
    'mean': F_mean_3d,
    'median': F_median_3d,
    'peak': F_peak_3d,
    'de': F_de_3d
}

print("======================================================================")
print("🧠 12개 단독 채널의 4가지 특징 변수 (총 48개) 개별 변수로 추출 및 저장")
print("  (Log Ratio 정규화된 특징 기반)")
print("======================================================================")

# 저장된 변수 이름을 추적할 임시 리스트
created_feature_names = []

# 모든 채널에 대해 반복
for ch_name in CHANNELS:
    try:
        ch_idx = ch_to_idx[ch_name]
    except KeyError:
        print(f"경고: 채널 '{ch_name}'가 CHANNELS 목록에 없습니다. 건너뜁니다.")
        continue
    
    # 4가지 특징 유형에 대해 반복
    for feat_type in FEATURE_TYPES:
        # 특징 배열 추출 (Log Ratio 차분 결과)
        feature_3d = feat_arrays[feat_type]
        
        # 3D 배열에서 해당 채널 인덱스만 슬라이싱하여 3차원 배열 유지
        # shape: (N_epochs, 1, N_bands)
        single_channel_3d = feature_3d[:, ch_idx:ch_idx+1, :]
        
        # 변수 이름 생성 (예: F_mean_3d_Fz)
        var_name = f"F_{feat_type}_3d_{ch_name}"
        
        # ⭐️ globals()를 사용하여 변수 이름으로 동적 할당
        globals()[var_name] = single_channel_3d
        
        created_feature_names.append(var_name)
        print(f"✅ 생성됨: {var_name}: {single_channel_3d.shape}")

print("----------------------------------------------------------------------")
print(f"총 생성된 개별 특징 변수 개수: {len(created_feature_names)}개 (12채널 * 4특징)")

# ⚠️ 예시 변수는 F_mean_3d, F_de_3d 등이 실제로 존재할 때만 출력 가능합니다.
# if 'F_mean_3d_Fz' in globals():
#     print(f"예시 변수 (Fz 채널의 Mean) shape: {globals()['F_mean_3d_Fz'].shape}")
# if 'F_de_3d_C3' in globals():
#     print(f"예시 변수 (C3 채널의 DE) shape: {globals()['F_de_3d_C3'].shape}")
```

```python
import numpy as np
from scipy.signal import welch

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 이미 정의되어야 합니다.
# X: (n_epochs, 12, n_samp) - 원본 12채널 EEG 데이터
# sfreq: 샘플링 주파수 (Hz)
#
# [이전 베이스라인 코드에서 생성된 12채널 특징 변수들]
# F_mean_baseline_matched: (n_epochs, 12, n_bands)
# F_median_baseline_matched: (n_epochs, 12, n_bands)
# F_peak_baseline_matched: (n_epochs, 12, n_bands)
# F_de_baseline_matched: (n_epochs, 12, n_bands)
# =========================================================================

# -------------------------------------------------------------------------
# 1. 주파수 밴드 설정
# -------------------------------------------------------------------------
BANDS = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
band_keys = list(BANDS.keys())
eps_stability = 1e-12

# --- 함수: 이벤트 기간 Log-PSD 및 DE 특징 계산 (Log Ratio 정규화 방식에 맞춤) ---
def compute_event_features_3d(X, sfreq, bands):
    """
    이벤트 기간(X)에 대해 4가지 특징(Log-PSD Mean, Median, Peak 및 DE)을 계산합니다.
    """
    n_ep, n_ch, n_samp = X.shape
    n_b = len(bands)
    nperseg = int(sfreq)
    
    # 결과 배열 초기화: 4가지 특징
    F_mean_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_median_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_peak_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_de_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)

    for i in range(n_ep):
        for ch in range(n_ch):
            channel_data = X[i, ch, :]
            
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12
            df = f[1] - f[0] if len(f) > 1 else 1.0

            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                idx = np.logical_and(f >= low, f <= high)
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    # 1. Log-PSD 기반 특징
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_event[i, ch, b_idx] = np.mean(band_log_psd)
                    F_median_event[i, ch, b_idx] = np.median(band_log_psd)
                    F_peak_event[i, ch, b_idx] = np.max(band_log_psd)
                    
                    # 2. DE 기반 특징
                    band_power = np.sum(band_psd) * df
                    if band_power > 0:
                        F_de_event[i, ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability)
                    else:
                        F_de_event[i, ch, b_idx] = 0.0
                else:
                    # 데이터가 없는 경우 0으로 처리
                    F_mean_event[i, ch, b_idx] = 0.0
                    F_median_event[i, ch, b_idx] = 0.0
                    F_peak_event[i, ch, b_idx] = 0.0
                    F_de_event[i, ch, b_idx] = 0.0
                    
    return F_mean_event, F_median_event, F_peak_event, F_de_event


# ==============================================================================
# 🌟 T7/T8/Fp1/Fp2 제외 (8ch) 및 Log Ratio 특징 계산 🌟
# ==============================================================================

# ░░ 2. T7/T8/Fp1/Fp2 채널 제외 및 데이터 재구성 ░░
CHANNELS_original = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2']
EXCLUDED_CHANNELS_8rm = ['T7', 'T8', 'Fp1', 'Fp2'] # 제외할 4개 채널
CHANNELS_8rm = [ch for ch in CHANNELS_original if ch not in EXCLUDED_CHANNELS_8rm]

# 제외할 채널의 인덱스 확인
ch_to_idx_original = {ch_name: i for i, ch_name in enumerate(CHANNELS_original)}
excluded_indices_8rm = [ch_to_idx_original[ch] for ch in EXCLUDED_CHANNELS_8rm]
all_indices = np.arange(len(CHANNELS_original))
retained_indices_8rm = np.delete(all_indices, excluded_indices_8rm)

# T7/T8/Fp1/Fp2을 제외한 8개 채널 데이터 (X_8rm)
X_8rm = X[:, retained_indices_8rm, :]

# T7/T8/Fp1/Fp2을 제외한 8개 채널 베이스라인 특징 (F_..._baseline_matched_8rm) 추출
# 베이스라인 특징 변수는 12채널 기준으로 존재한다고 가정하고 슬라이싱합니다.
F_mean_baseline_matched_8rm = F_mean_baseline_matched[:, retained_indices_8rm, :]
F_median_baseline_matched_8rm = F_median_baseline_matched[:, retained_indices_8rm, :]
F_peak_baseline_matched_8rm = F_peak_baseline_matched[:, retained_indices_8rm, :]
F_de_baseline_matched_8rm = F_de_baseline_matched[:, retained_indices_8rm, :]


# ░░ 3. 8채널 데이터로 이벤트 특징 계산 ░░
print("\n[Start] T7/T8/Fp1/Fp2 제외 (8ch, _8rm) 이벤트 특징 계산 시작...")
F_mean_event_8rm, F_median_event_8rm, F_peak_event_8rm, F_de_event_8rm = \
    compute_event_features_3d(X_8rm, sfreq, BANDS)
print("[Done] T7/T8/Fp1/Fp2 제외 (8ch, _8rm) 이벤트 특징 계산 완료.")


# ░░ 4. Log Ratio 정규화 (차분 방식) ░░
# F_norm = F_event - F_baseline
print("[Start] Log Ratio 정규화 (차분) 적용 중...")
F_mean_3d_8rm = F_mean_event_8rm - F_mean_baseline_matched_8rm
F_median_3d_8rm = F_median_event_8rm - F_median_baseline_matched_8rm
F_peak_3d_8rm = F_peak_event_8rm - F_peak_baseline_matched_8rm
F_de_3d_8rm = F_de_event_8rm - F_de_baseline_matched_8rm
print("[Done] Log Ratio 정규화 완료.")


# ░░ 5. 비대칭 특징 (DASM, RASM) 계산 (T7/T8/Fp1/Fp2 제외) ░░

# 5-1. 채널 정보 및 비대칭 쌍 정의 (3쌍만 남음)
HEM_PAIRS_8rm = [
    ('F3', 'F4'), ('F7', 'F8'), ('C3', 'C4') 
]
pair_names_8rm = [f"{L}-{R}" for L, R in HEM_PAIRS_8rm]

# 새로운 채널 목록에 맞춘 인덱스 딕셔너리 생성
ch_to_idx_8rm = {ch_name: i for i, ch_name in enumerate(CHANNELS_8rm)}

# 5-2. 특징 배열 초기화
n_ep_8rm, n_ch_8rm, n_b_8rm = F_de_3d_8rm.shape
n_pairs_8rm = len(HEM_PAIRS_8rm)
F_dasm_3d_8rm = np.empty((n_ep_8rm, n_pairs_8rm, n_b_8rm), dtype=np.float32)
F_rasm_3d_8rm = np.empty((n_ep_8rm, n_pairs_8rm, n_b_8rm), dtype=np.float32) 


# 5-3. 각 채널 쌍에 대해 DASM_8rm, RASM_8rm 계산 (정규화된 DE 특징 사용)
for i, (left_ch, right_ch) in enumerate(HEM_PAIRS_8rm):
    left_idx = ch_to_idx_8rm[left_ch]
    right_idx = ch_to_idx_8rm[right_ch]
    
    # 정규화된 DE 값 추출 (DE 차분 결과)
    de_norm_left = F_de_3d_8rm[:, left_idx, :]
    de_norm_right = F_de_3d_8rm[:, right_idx, :]
    
    # DASM 계산: 정규화된 DE의 차분
    F_dasm_3d_8rm[:, i, :] = de_norm_left - de_norm_right
    
    # RASM 계산: 정규화된 DE의 비율 (분모에 안정성 항 추가)
    F_rasm_3d_8rm[:, i, :] = de_norm_left / (de_norm_right + eps_stability)


# ==============================================================================
# ░░ 6. 최종 결과 확인 및 요약 ░░
# ==============================================================================

print("\n================================================================")
print(" ✅ T7/T8/Fp1/Fp2 채널 제외 (8ch, _8rm) Log-PSD/DE 기반 특징 변수 생성 완료")
print("================================================================")

# 1. 8 채널 기반 특징 변수
print("### 1. 8 채널 기반 기본 특징 (Log-PSD/DE 차분)")
print(f"  - 채널 목록 (8ch): {CHANNELS_8rm}")
print(f"  - F_mean_3d_8rm (Log-PSD Mean 차분): {F_mean_3d_8rm.shape}")
print(f"  - F_median_3d_8rm (Log-PSD Median 차분): {F_median_3d_8rm.shape}")
print(f"  - F_peak_3d_8rm (Log-PSD Peak 차분): {F_peak_3d_8rm.shape}")
print(f"  - F_de_3d_8rm (DE 차분): {F_de_3d_8rm.shape}")

print("\n### 2. 8 채널 기반 비대칭 특징 (DASM/RASM)")
print(f"  - 채널 쌍 목록 (3쌍): {pair_names_8rm}")
print(f"  - F_dasm_3d_8rm (정규화된 DE의 차분): {F_dasm_3d_8rm.shape}")
print(f"  - F_rasm_3d_8rm (정규화된 DE의 비율): {F_rasm_3d_8rm.shape}")

print(f"\n참고: Epoch 수 = {F_de_3d_8rm.shape[0]}, 주파수 밴드 수 = {F_de_3d_8rm.shape[2]}")
print("================================================================")
```

```python
import numpy as np
from scipy.signal import welch

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 이미 정의되어야 합니다.
# X: (n_epochs, 12, n_samp) - 원본 12채널 EEG 데이터
# sfreq: 샘플링 주파수 (Hz)
# y: (n_epochs,) - 레이블 (최종 출력에 사용)
#
# [이전 베이스라인 코드에서 생성된 12채널 특징 변수들]
# F_mean_baseline_matched: (n_epochs, 12, n_bands)
# F_median_baseline_matched: (n_epochs, 12, n_bands)
# F_peak_baseline_matched: (n_epochs, 12, n_bands)
# F_de_baseline_matched: (n_epochs, 12, n_bands)
# =========================================================================

# ░░ 1. 주파수 밴드 설정 ░░
BANDS = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
band_keys = list(BANDS.keys())
eps_stability = 1e-12

# ░░ 2. 채널 영역 정의 및 인덱스 변환 ░░
AREAS = {
    "Area1_Frontal": ['Fz', 'F3', 'F4', 'F7', 'F8', 'Fp1', 'Fp2'], # 전두엽/전전두엽
    "Area2_Central": ['Cz', 'C3', 'C4', 'T7', 'T8']                  # 중앙/측두엽
}
area_names = list(AREAS.keys())

# 안전을 위해 12채널 목록을 가정하여 정의합니다.
try:
    _ = CHANNELS
except NameError:
    CHANNELS = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2']
    
# 채널 이름을 인덱스로 변환하는 딕셔너리 생성
ch_to_idx = {ch_name: i for i, ch_name in enumerate(CHANNELS)}

# 영역별 채널 인덱스 목록
AREA_INDICES = {
    area: [ch_to_idx[ch] for ch in ch_list if ch in ch_to_idx]
    for area, ch_list in AREAS.items()
}

# ░░ 3. 이벤트 기간의 Log-PSD 및 DE 특징 계산 함수 (영역 평균) ░░
def compute_event_features_2area(X, sfreq, bands, area_indices):
    """
    X: (n_epochs, n_ch, n_samp)
    반환: F_mean_event_2area, F_median_event_2area, F_peak_event_2area, F_de_event_2area
          (4개의 (n_epochs, n_areas, n_bands) 배열)
    """
    n_ep, n_ch, n_samp = X.shape
    n_b = len(bands)
    n_a = len(area_indices)
    nperseg = int(sfreq)
    
    # 결과 배열 초기화
    F_mean_event_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    F_median_event_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    F_peak_event_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    F_de_event_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    
    for i in range(n_ep):
        epoch_data = X[i]  # (n_ch, n_samp)
        
        # 1. 에포크 내 모든 채널에 대해 4가지 특징을 모두 계산
        F_mean_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        F_median_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        F_peak_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        F_de_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        
        for ch in range(n_ch):
            channel_data = epoch_data[ch, :]
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12
            df = f[1] - f[0] if len(f) > 1 else 1.0

            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                idx = np.logical_and(f >= low, f <= high)
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_epoch[ch, b_idx] = np.mean(band_log_psd)
                    F_median_epoch[ch, b_idx] = np.median(band_log_psd)
                    F_peak_epoch[ch, b_idx] = np.max(band_log_psd)
                    
                    band_power = np.sum(band_psd) * df
                    F_de_epoch[ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability) if band_power > 0 else 0.0
                else:
                    F_mean_epoch[ch, b_idx] = 0.0
                    F_median_epoch[ch, b_idx] = 0.0
                    F_peak_epoch[ch, b_idx] = 0.0
                    F_de_epoch[ch, b_idx] = 0.0
                    
        # 2. 영역별 평균 특징 계산
        for a_idx, (area, indices) in enumerate(area_indices.items()):
            if indices:
                # Log-PSD Mean 특징의 영역 평균
                F_mean_event_2area[i, a_idx, :] = np.mean(F_mean_epoch[indices, :], axis=0)
                # Log-PSD Median 특징의 영역 평균
                F_median_event_2area[i, a_idx, :] = np.mean(F_median_epoch[indices, :], axis=0)
                # Log-PSD Peak 특징의 영역 평균
                F_peak_event_2area[i, a_idx, :] = np.mean(F_peak_epoch[indices, :], axis=0)
                # DE 특징의 영역 평균
                F_de_event_2area[i, a_idx, :] = np.mean(F_de_epoch[indices, :], axis=0)
            else:
                F_mean_event_2area[i, a_idx, :] = 0.0
                F_median_event_2area[i, a_idx, :] = 0.0
                F_peak_event_2area[i, a_idx, :] = 0.0
                F_de_event_2area[i, a_idx, :] = 0.0
                        
    return F_mean_event_2area, F_median_event_2area, F_peak_event_2area, F_de_event_2area

# ░░ 4. 이벤트 기간의 영역 평균 특징 추출 ░░
print("\n[Start] 12채널 2영역 이벤트 특징 (Log-PSD/DE) 계산 시작...")
F_mean_event_2area, F_median_event_2area, F_peak_event_2area, F_de_event_2area = \
    compute_event_features_2area(X, sfreq, BANDS, AREA_INDICES)
print("[Done] 12채널 2영역 이벤트 특징 계산 완료.")


# ░░ 5. 베이스라인 특징 추출 (영역 평균) ░░
# F_..._baseline_matched: (n_epochs, n_ch_original, n_bands)에서 영역 평균 계산

n_ep = F_mean_baseline_matched.shape[0]
n_b = len(BANDS)
n_a = len(AREA_INDICES)

# 영역별 베이스라인 특징 평균을 저장할 배열 초기화
F_mean_baseline_matched_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
F_median_baseline_matched_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
F_peak_baseline_matched_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
F_de_baseline_matched_2area = np.empty((n_ep, n_a, n_b), dtype=np.float32)


for a_idx, (area, indices) in enumerate(AREA_INDICES.items()):
    if indices:
        # 해당 영역 채널들의 각 특징 평균을 계산 (axis=1: 채널 차원)
        F_mean_baseline_matched_2area[:, a_idx, :] = np.mean(F_mean_baseline_matched[:, indices, :], axis=1)
        F_median_baseline_matched_2area[:, a_idx, :] = np.mean(F_median_baseline_matched[:, indices, :], axis=1)
        F_peak_baseline_matched_2area[:, a_idx, :] = np.mean(F_peak_baseline_matched[:, indices, :], axis=1)
        F_de_baseline_matched_2area[:, a_idx, :] = np.mean(F_de_baseline_matched[:, indices, :], axis=1)
    else:
        # 채널이 없는 경우 0으로 처리 (안정성 항 불필요)
        F_mean_baseline_matched_2area[:, a_idx, :] = 0.0
        F_median_baseline_matched_2area[:, a_idx, :] = 0.0
        F_peak_baseline_matched_2area[:, a_idx, :] = 0.0
        F_de_baseline_matched_2area[:, a_idx, :] = 0.0


# ░░ 6. Log Ratio 정규화 (차분 방식) ░░
# F_norm = F_event - F_baseline
print("[Start] Log Ratio 정규화 (차분) 적용 중...")
F_mean_3d_2area = F_mean_event_2area - F_mean_baseline_matched_2area
F_median_3d_2area = F_median_event_2area - F_median_baseline_matched_2area
F_peak_3d_2area = F_peak_event_2area - F_peak_baseline_matched_2area
F_de_3d_2area = F_de_event_2area - F_de_baseline_matched_2area
print("[Done] Log Ratio 정규화 완료.")

# ------------------------------------------------------------------------------
# 🌟 비대칭 특징 (DASM, RASM) 계산 🌟
# 비대칭 특징은 보통 좌/우 반구의 차이를 계산하므로,
# Area1 (Frontal)의 좌/우 채널을 기반으로 할 수 있지만,
# 현재 영역(Area1: Frontal, Area2: Central)은 좌/우 대칭 쌍이 아니므로
# **영역 간의 차분 특징**을 계산하여 유사한 대칭 특징으로 간주합니다.
# ------------------------------------------------------------------------------
n_ep_2area, n_a_2area, n_b_2area = F_de_3d_2area.shape

# F_dasm_3d_2area (영역 차분): Area1 - Area2
# (n_epochs, 1, n_bands)로 생성
F_dasm_3d_2area = np.empty((n_ep_2area, 1, n_b_2area), dtype=np.float32)
F_dasm_3d_2area[:, 0, :] = F_de_3d_2area[:, 0, :] - F_de_3d_2area[:, 1, :]

# F_rasm_3d_2area (영역 비율): Area1 / Area2
# (n_epochs, 1, n_bands)로 생성
F_rasm_3d_2area = np.empty((n_ep_2area, 1, n_b_2area), dtype=np.float32)
F_rasm_3d_2area[:, 0, :] = F_de_3d_2area[:, 0, :] / (F_de_3d_2area[:, 1, :] + eps_stability)


# ░░ 7. 최종 결과 확인 ░░
print("\n================================================================")
print(" ✅ 12채널 2영역 Log-PSD/DE 기반 특징 변수 생성 완료 (차분 기반)")
print("================================================================")
print("### 1. 영역별 특징 (2영역)")
print(f"  - 영역: {area_names}")
print(f"  - F_mean_3d_2area (Log-PSD Mean 차분): {F_mean_3d_2area.shape}")
print(f"  - F_median_3d_2area (Log-PSD Median 차분): {F_median_3d_2area.shape}")
print(f"  - F_peak_3d_2area (Log-PSD Peak 차분): {F_peak_3d_2area.shape}")
print(f"  - F_de_3d_2area (DE 차분): {F_de_3d_2area.shape}")

print("\n### 2. 영역 간 비대칭 특징 (1쌍)")
print("  - 쌍: Area1_Frontal - Area2_Central")
print(f"  - F_dasm_3d_2area (DE 차분 차이): {F_dasm_3d_2area.shape}")
print(f"  - F_rasm_3d_2area (DE 차분 비율): {F_rasm_3d_2area.shape}")
print(f"\n참고: Epoch 수 = {F_de_3d_2area.shape[0]}, 주파수 밴드 수 = {len(BANDS)}")
print("================================================================")
```

```python
import numpy as np
from scipy.signal import welch

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 이미 정의되어야 합니다.
# X: (n_epochs, 12, n_samp) - 원본 12채널 EEG 데이터
# sfreq: 샘플링 주파수 (Hz)
# y: (n_epochs,) - 레이블 (최종 출력에 사용)
#
# [이전 베이스라인 코드에서 생성된 12채널 특징 변수들]
# F_mean_baseline_matched: (n_epochs, 12, n_bands)
# F_median_baseline_matched: (n_epochs, 12, n_bands)
# F_peak_baseline_matched: (n_epochs, 12, n_bands)
# F_de_baseline_matched: (n_epochs, 12, n_bands)
# P_baseline_matched: (n_epochs, 12, n_bands) - (사용하지 않지만, 기존 코드 호환을 위해 명시)
# =========================================================================

# ░░ 1. 주파수 밴드 설정 ░░
BANDS = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
band_keys = list(BANDS.keys())
eps_stability = 1e-12

# 안전을 위해 12채널 목록을 가정하여 정의합니다.
try:
    _ = CHANNELS
except NameError:
    CHANNELS = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2']

# ░░ 2. 채널 영역 정의 및 인덱스 변환 (3개 영역) ░░
AREAS_3area = {
    "Area1_Left": ['Fp1', 'F3', 'C3', 'F7', 'T7'],     # 좌뇌 영역 (5채널)
    "Area2_Right": ['Fp2', 'F4', 'F8', 'C4', 'T8'],    # 우뇌 영역 (5채널)
    "Area3_Midline": ['Fz', 'Cz']                      # 정중선 영역 (2채널)
}
area_names_3area = list(AREAS_3area.keys())

# 채널 이름을 인덱스로 변환하는 딕셔너리 생성
ch_to_idx = {ch_name: i for i, ch_name in enumerate(CHANNELS)}

# 영역별 채널 인덱스 목록
AREA_INDICES_3area = {
    area: [ch_to_idx[ch] for ch in ch_list if ch in ch_to_idx]
    for area, ch_list in AREAS_3area.items()
}

# ░░ 3. 이벤트 기간의 Log-PSD 및 DE 특징 계산 함수 (영역 평균) ░░
def compute_event_features_3area(X, sfreq, bands, area_indices):
    """
    X: (n_epochs, n_ch, n_samp)
    반환: F_mean_event_3area, F_median_event_3area, F_peak_event_3area, F_de_event_3area
          (4개의 (n_epochs, n_areas, n_bands) 배열)
    """
    n_ep, n_ch, n_samp = X.shape
    n_b = len(bands)
    n_a = len(area_indices)
    nperseg = int(sfreq)
    
    # 결과 배열 초기화
    F_mean_event_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    F_median_event_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    F_peak_event_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    F_de_event_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
    
    for i in range(n_ep):
        epoch_data = X[i]  # (n_ch, n_samp)
        
        # 1. 에포크 내 모든 채널에 대해 4가지 특징을 모두 계산
        F_mean_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        F_median_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        F_peak_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        F_de_epoch = np.empty((n_ch, n_b), dtype=np.float32)
        
        for ch in range(n_ch):
            channel_data = epoch_data[ch, :]
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12
            df = f[1] - f[0] if len(f) > 1 else 1.0

            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                idx = np.logical_and(f >= low, f <= high)
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    # Log-PSD 기반 특징 계산
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_epoch[ch, b_idx] = np.mean(band_log_psd)
                    F_median_epoch[ch, b_idx] = np.median(band_log_psd)
                    F_peak_epoch[ch, b_idx] = np.max(band_log_psd)
                    
                    # DE 특징 계산
                    band_power = np.sum(band_psd) * df
                    F_de_epoch[ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability) if band_power > 0 else 0.0
                else:
                    F_mean_epoch[ch, b_idx] = 0.0
                    F_median_epoch[ch, b_idx] = 0.0
                    F_peak_epoch[ch, b_idx] = 0.0
                    F_de_epoch[ch, b_idx] = 0.0
                    
        # 2. 영역별 평균 특징 계산
        for a_idx, (area, indices) in enumerate(area_indices.items()):
            if indices:
                # Log-PSD/DE 특징의 영역 평균을 계산
                F_mean_event_3area[i, a_idx, :] = np.mean(F_mean_epoch[indices, :], axis=0)
                F_median_event_3area[i, a_idx, :] = np.mean(F_median_epoch[indices, :], axis=0)
                F_peak_event_3area[i, a_idx, :] = np.mean(F_peak_epoch[indices, :], axis=0)
                F_de_event_3area[i, a_idx, :] = np.mean(F_de_epoch[indices, :], axis=0)
            else:
                F_mean_event_3area[i, a_idx, :] = 0.0
                F_median_event_3area[i, a_idx, :] = 0.0
                F_peak_event_3area[i, a_idx, :] = 0.0
                F_de_event_3area[i, a_idx, :] = 0.0
                        
    return F_mean_event_3area, F_median_event_3area, F_peak_event_3area, F_de_event_3area


# ░░ 4. 이벤트 기간의 3영역 평균 특징 추출 ░░
print("\n[Start] 12채널 3영역 이벤트 특징 (Log-PSD/DE) 계산 시작...")
F_mean_event_3area, F_median_event_3area, F_peak_event_3area, F_de_event_3area = \
    compute_event_features_3area(X, sfreq, BANDS, AREA_INDICES_3area)
print("[Done] 12채널 3영역 이벤트 특징 계산 완료.")


# ░░ 5. 베이스라인 특징 추출 (3영역 평균) ░░

n_ep = F_mean_baseline_matched.shape[0]
n_b = len(BANDS)
n_a = len(AREAS_3area)

# 영역별 베이스라인 특징 평균을 저장할 배열 초기화
F_mean_baseline_matched_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
F_median_baseline_matched_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
F_peak_baseline_matched_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)
F_de_baseline_matched_3area = np.empty((n_ep, n_a, n_b), dtype=np.float32)

for a_idx, (area, indices) in enumerate(AREA_INDICES_3area.items()):
    if indices:
        # 해당 영역 채널들의 각 특징 평균을 계산 (axis=1: 채널 차원)
        F_mean_baseline_matched_3area[:, a_idx, :] = np.mean(F_mean_baseline_matched[:, indices, :], axis=1)
        F_median_baseline_matched_3area[:, a_idx, :] = np.mean(F_median_baseline_matched[:, indices, :], axis=1)
        F_peak_baseline_matched_3area[:, a_idx, :] = np.mean(F_peak_baseline_matched[:, indices, :], axis=1)
        F_de_baseline_matched_3area[:, a_idx, :] = np.mean(F_de_baseline_matched[:, indices, :], axis=1)
    else:
        # 채널이 없는 경우 0으로 처리
        F_mean_baseline_matched_3area[:, a_idx, :] = 0.0
        F_median_baseline_matched_3area[:, a_idx, :] = 0.0
        F_peak_baseline_matched_3area[:, a_idx, :] = 0.0
        F_de_baseline_matched_3area[:, a_idx, :] = 0.0


# ░░ 6. Log Ratio 정규화 (차분 방식) ░░
# F_norm = F_event - F_baseline
print("[Start] Log Ratio 정규화 (차분) 적용 중...")
F_mean_3d_3area = F_mean_event_3area - F_mean_baseline_matched_3area
F_median_3d_3area = F_median_event_3area - F_median_baseline_matched_3area
F_peak_3d_3area = F_peak_event_3area - F_peak_baseline_matched_3area
# 비대칭 특징 계산에 사용될 F_de_3d_3area도 차분으로 계산합니다.
F_de_3d_3area = F_de_event_3area - F_de_baseline_matched_3area
print("[Done] Log Ratio 정규화 완료.")


# ------------------------------------------------------------------------------
# 🌟 7. 비대칭 특징 (DASM, RASM) 계산 (DE 차분 기반) 🌟
# ------------------------------------------------------------------------------

# 7-1. 비대칭 특징 계산에 사용할 영역 인덱스 확인
left_area_name = "Area1_Left"
right_area_name = "Area2_Right"
    
try:
    left_idx_3area = area_names_3area.index(left_area_name)
    right_idx_3area = area_names_3area.index(right_area_name)

    # 특징 배열 초기화
    n_ep_3area, _, n_b_3area = F_de_3d_3area.shape
    
    # 영역쌍은 1개(Left/Right)
    F_dasm_3d_3area = np.empty((n_ep_3area, 1, n_b_3area), dtype=np.float32)
    F_rasm_3d_3area = np.empty((n_ep_3area, 1, n_b_3area), dtype=np.float32)

    # Left 및 Right 영역의 정규화된 DE 값 추출 (shape: n_epochs, n_bands)
    de_norm_left_area = F_de_3d_3area[:, left_idx_3area, :]
    de_norm_right_area = F_de_3d_3area[:, right_idx_3area, :]
    
    # DASM 계산: DE_Norm(좌뇌 영역) - DE_Norm(우뇌 영역)
    F_dasm_3d_3area[:, 0, :] = de_norm_left_area - de_norm_right_area
    
    # RASM 계산: DE_Norm(좌뇌 영역) / DE_Norm(우뇌 영역)
    # 분모에 안정성 항 추가
    F_rasm_3d_3area[:, 0, :] = de_norm_left_area / (de_norm_right_area + eps_stability)

    print("\n[Done] 3개 영역 기반 DASM/RASM 계산 완료 (DE 차분 기반).")
    
except ValueError as e:
    print(f"\n⚠️ 경고: DASM/RASM 계산 실패. 영역 이름 찾기 오류: {e}")
    F_dasm_3d_3area = np.empty((F_de_3d_3area.shape[0], 1, F_de_3d_3area.shape[2]), dtype=np.float32)
    F_rasm_3d_3area = np.empty((F_de_3d_3area.shape[0], 1, F_de_3d_3area.shape[2]), dtype=np.float32)


# ░░ 8. 최종 결과 확인 ░░
print("\n================================================================")
print(" ✅ 12채널 3영역 Log-PSD/DE 기반 특징 변수 (3D) 생성 완료 (차분 기반)")
print("================================================================")

# 1. 영역별 특징
print("### 1. 영역별 특징 (3영역)")
print(f"  - 영역: {area_names_3area}")
print(f"  - F_mean_3d_3area (Log-PSD Mean 차분): {F_mean_3d_3area.shape}")
print(f"  - F_median_3d_3area (Log-PSD Median 차분): {F_median_3d_3area.shape}")
print(f"  - F_peak_3d_3area (Log-PSD Peak 차분): {F_peak_3d_3area.shape}")
print(f"  - F_de_3d_3area (DE 차분): {F_de_3d_3area.shape}")

# 2. 비대칭 특징
print("\n### 2. 비대칭 특징 (1쌍: Left-Right)")
print(f"  - F_dasm_3d_3area (DE 차분 차이): {F_dasm_3d_3area.shape}")
print(f"  - F_rasm_3d_3area (DE 차분 비율): {F_rasm_3d_3area.shape}")

# y.shape[0]이 정의되어 있지 않거나 배열이 아닐 경우 오류 방지
try:
    print(f"\n참고: Epoch 수 = {y.shape[0]}, n_areas={len(area_names_3area)}, n_bands={len(BANDS)}")
except NameError:
    print("\n참고: Epoch 수: (변수 y 미정의), n_areas=3, n_bands=5")
print("================================================================")
```

```python
import mne
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd
from collections import Counter
import warnings

# 경고 메시지를 필터링하여 출력하지 않음
warnings.filterwarnings("ignore", category=DeprecationWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)
warnings.filterwarnings("ignore", category=UserWarning)

# 파일 경로 설정 (사용자 요청 경로)
file_path = r"1018 승욱 후각.bdf" # 또는 .bdf 파일

# 파일 확장자에 따라 적절한 함수 사용
file_extension = os.path.splitext(file_path)[1].lower()

# --- 파일 로드 ---
try:
    if file_extension == '.bdf':
        raw2 = mne.io.read_raw_bdf(file_path, preload=True, verbose=False) # raw 대신 raw2 사용
        print("BDF 파일이 로드되었습니다.")
    elif file_extension == '.set':
        raw2 = mne.io.read_raw_eeglab(file_path, preload=True, verbose=False) # raw 대신 raw2 사용
        print("SET 파일이 로드되었습니다.")
    else:
        raise ValueError(f"지원되지 않는 파일 형식입니다: {file_extension}")
except FileNotFoundError:
    print(f"오류: 파일을 찾을 수 없습니다: {file_path}")
    # 파일 로드 실패 시 나머지 코드가 실행되지 않도록 종료
    exit()

# ==========================================================
# ✅ [1.5] T3/T4 채널 이름 강제 수정 (위치 오류 해결)
# 몽타주 좌표 인식을 위해 T3/T4를 T7/T8로 수정합니다.
# ==========================================================
mapping = {}
if 'T3' in raw2.ch_names:
    raw2.rename_channels({'T3': 'T7'})
    mapping['T3'] = 'T7'
if 'T4' in raw2.ch_names:
    raw2.rename_channels({'T4': 'T8'})
    mapping['T4'] = 'T8'

if mapping:
    print(f"채널 이름 수정: {mapping} (몽타주 위치 정확도 향상)")
# ==========================================================


print("모든 채널 목록:", raw2.ch_names)
print(f"샘플링 주파수: {raw2.info['sfreq']} Hz, 총 길이: {raw2.n_times/raw2.info['sfreq']:.1f} s")

# --- 채널 타입 정리(옵션): EOG로 지정하면 나중에 picks로 쉽게 제외 가능 ---
type_map = {}
for cand in ['ExG 1','ExG 2','EOG','EOG1','EOG2']:
    if cand in raw2.ch_names:
        type_map[cand] = 'eog'
if type_map:
    raw2.set_channel_types(type_map) # raw 대신 raw2 사용

# --- 불필요 채널 제거 ---
channels_to_remove = ['ACC22','ACC23','ACC24','Packet Counter']  # 가속도나 패킷 카운터 등
# ch 대신 ch2 사용
existing_channels_to_remove = [ch2 for ch2 in channels_to_remove if ch2 in raw2.ch_names] 
if existing_channels_to_remove:
    raw2.drop_channels(existing_channels_to_remove) # raw 대신 raw2 사용
    print(f"제거된 채널: {existing_channels_to_remove}")


# --- 몽타주: cap에 맞게 선택 및 채널 위치 시각화 ---
montage_set_successfully = False
try:
    # BioSemi 특유 이름이 있는지 확인
    is_biosemi = any(name.startswith('A') or name.startswith('B') for name in raw2.ch_names) and len(raw2.ch_names) > 32 # raw 대신 raw2 사용

    if is_biosemi:
        montage = mne.channels.make_standard_montage('biosemi64')
        print("BioSemi 64-channel 몽타주를 시도합니다.")
    else:
        # 32채널 이하이거나 표준 명명법일 경우
        montage = mne.channels.make_standard_montage('standard_1020')
        print("Standard 10-20 몽타주를 시도합니다.")
    
    # 몽타주 적용
    raw2.set_montage(montage, on_missing='warn') # raw 대신 raw2 사용
    print("몽타주가 성공적으로 설정되었습니다.")
    montage_set_successfully = True

except Exception as e:
    print(f"몽타주 설정 경고/오류: {e}. 채널 위치 시각화가 어려울 수 있습니다.")


# ==========================================================
# ✅ 채널 위치 시각화
# ==========================================================
if montage_set_successfully:
    try:
        # 2D 평면 위에 채널 위치(topomap)를 플롯
        raw2.plot_sensors( # raw 대신 raw2 사용
            kind='topomap',
            title='EEG Channel Locations (Topomap) - All Channels',
            show_names=True,
            sphere='auto'
        )
        print("\n[채널 위치 시각화] Topomap 플롯이 생성되었습니다.")
    except Exception as e:
        print(f"\n[경고] 채널 위치 시각화 실패: {e}")
else:
    print("\n[알림] 몽타주 설정 실패로 채널 위치 시각화는 건너뜁니다.")
# ==========================================================


# --- 이벤트 추출 함수 ---
def extract_events_safe(raw2, file_ext): # raw 대신 raw2 사용
    # 1) EEGLAB: annotations 우선
    if file_ext == '.set':
        if len(raw2.annotations) > 0: # raw 대신 raw2 사용
            ev, ev_id = mne.events_from_annotations(raw2) # raw 대신 raw2 사용
            print(f"SET/Annotations 기반 이벤트: {len(ev)}개, event_id={ev_id}")
            return ev
        else:
            print("경고: SET에 annotations가 없습니다. stim 채널 탐색을 시도합니다.")

    # 2) BDF 혹은 stim 채널이 있는 경우: 자극 채널 이름 자동 탐색
    stim_name = None
    for cand in ('Status', 'STI 014', 'TRIGGER', 'Stim', 'STIM'):
        if cand in raw2.ch_names: # raw 대신 raw2 사용
            stim_name = cand
            break
    if stim_name is None:
        stim_picks = mne.pick_types(raw2.info, stim=True) # raw 대신 raw2 사용
        if len(stim_picks) > 0:
            stim_name = raw2.ch_names[stim_picks[0]] # raw 대신 raw2 사용

    if stim_name is None:
        raise RuntimeError("자극(stim) 채널을 찾을 수 없습니다. (.set이면 annotations 확인 필요)")

    ev = mne.find_events(
        raw2, # raw 대신 raw2 사용
        stim_channel=stim_name,
        shortest_event=1,
        consecutive='increasing',
        mask=0xFF,         # 하위 8비트만 사용 (BioSemi 권장 패턴)
        mask_type='and',
        verbose=False
    )
    print(f"STIM 채널('{stim_name}') 기반 이벤트: {len(ev)}개")
    return ev

try:
    events2 = extract_events_safe(raw2, file_extension) # raw 대신 raw2 사용
except Exception as e:
    print(f"이벤트 추출 실패: {e}")
    events2 = np.empty((0,3), dtype=int)

# --- 이벤트 요약 출력 및 블록 길이 계산 ---
if len(events2) > 0:
    codes = events2[:,2]
    print(f"\n추출된 이벤트 수: {len(events2)}")
    print("코드 분포:", dict(Counter(codes)))
    
    print("\n--- 이벤트 상세 정보 (시점 및 코드) ---")
    sfreq2 = raw2.info['sfreq'] # raw 대신 raw2 사용
    for event in events2:
        sample = event[0]
        event_code = event[2]
        timestamp_sec = sample / sfreq2
        print(f"시점: {timestamp_sec:.2f}s | 샘플: {sample} | 코드: {event_code}")

    # 1(rest) 이벤트의 시작점 찾기
    rest_idx = np.where(events2[:, 2] == 1)[0]
    
    print("\n--- Task 블록 길이 (Rest-to-Rest) ---")
    # i 대신 i2 사용
    for i2 in range(len(rest_idx) - 1):
        # 현재 rest 다음 이벤트(task 시작)부터 다음 rest 이벤트(task 끝)까지
        start_sample = events2[rest_idx[i2] + 1][0]
        end_sample = events2[rest_idx[i2+1]][0]
        duration_sec = (end_sample - start_sample) / sfreq2
        block_label = events2[rest_idx[i2] + 1][2] if rest_idx[i2] + 1 < len(events2) else "N/A"
        print(f"블록 {i2+1} (코드 {block_label}): {duration_sec:.2f}초")
    
    print("\n--- Task to Task 블록 길이 (Task_Start-to-Next_Task_Start) ---")
    # i 대신 i2 사용
    for i2 in range(len(rest_idx) - 1):
        try:
            # 현재 rest 다음 이벤트(task 시작)
            start_sample = events2[rest_idx[i2] + 1][0]
            # 다음 rest 다음 이벤트(다음 task 시작)
            end_sample = events2[rest_idx[i2+1] + 1][0]
            duration_sec = (end_sample - start_sample) / sfreq2
            
            # 'Task-to-Task' 블록의 첫 번째 이벤트 라벨
            block_label = events2[rest_idx[i2] + 1][2]
            print(f"Task to Task 블록 {i2+1} (코드 {block_label}): {duration_sec:.2f}초")
        except IndexError:
            # 마지막 rest 이후에 task가 없어 IndexError 발생 시
            print(f"경고: Task to Task 블록 {i2+1}의 끝점을 찾을 수 없습니다. (마지막 블록)")
            continue
    
    # (옵션) 이벤트 시각화
    mne.viz.plot_events(events2, sfreq=raw2.info['sfreq']) # raw 대신 raw2 사용
else:
    print("\n추출된 이벤트가 없습니다.")


# ==========================================================
# ✅ [2] 선택 채널 전처리 및 시각화 (12 Channels)
# ==========================================================
print("\n" + "="*50)
print("--- [2] 선택 채널 전처리 (12 Channels) ---")

# T3->T7, T4->T8로 이름이 수정되었으므로, 선택 채널 리스트도 T7, T8을 사용합니다.
selected_channels_12 = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2']
raw_channels = raw2.ch_names # raw 대신 raw2 사용

# 실제 데이터에 존재하는 채널만 필터링 (오류 방지)
existing_selected_channels_12 = [ch for ch in selected_channels_12 if ch in raw_channels]

if not existing_selected_channels_12:
    print(f"\n경고: 요청하신 12개 채널 중 데이터에 존재하는 채널이 없습니다. 전처리를 건너뜁니다.")
    print(f"요청 채널: {selected_channels_12}")
    # 12채널 데이터가 없으므로 raw_selected2를 정의하지 않습니다.
else:
    if len(existing_selected_channels_12) < len(selected_channels_12):
        print(f"\n경고: 요청하신 12개 채널 중 {len(selected_channels_12) - len(existing_selected_channels_12)}개가 데이터에 없어 제외되었습니다.")
        print(f"실제 전처리 채널: {existing_selected_channels_12}")

    # 선택한 채널만 복사해서 새 객체 raw_selected2로 만들기
    raw_selected2 = raw2.copy().pick_channels(existing_selected_channels_12) # raw 대신 raw2 사용
    
    # 몽타주 시각화
    if montage_set_successfully:
        try:
            # 선택된 채널의 위치 시각화
            raw_selected2.plot_sensors(
                show_names=True, 
                kind='topomap', 
                sphere='auto',
                title='Selected 12 Channel Locations (T7/T8 Renamed)'
            )
        except Exception as e:
            print(f"\n[경고] 선택 채널 위치 시각화 실패: {e}")
    else:
        print("\n[알림] 원본 raw2에 몽타주가 설정되지 않아 선택 채널 위치 시각화를 건너뜁니다.")


    # 필터 제거 전 PSD 시각화
    plt.figure(figsize=(10, 4))
    raw_selected2.plot_psd(fmax=100, show=False)
    plt.title('PSD Before FIR Filtering (Selected 12 Channels)')
    
    # FIR 대역 통과필터: 0.5Hz~ 55Hz
    print("\nraw_selected2 FIR 필터링 (0.5Hz ~ 55Hz) 시작...")
    raw_selected2.filter(
        l_freq=0.5,      # 저주파 차단 주파수
        h_freq=55.0,     # 고주파 차단 주파수
        method='fir',    # FIR 필터 사용
        fir_design='firwin', # FIR 설계 방식
        phase='zero',    # 위상 지연 보정 (양방향 필터링)
    )
    print("raw_selected2 FIR 필터링 완료.")

    # 필터 제거 후 PSD 시각화
    plt.figure(figsize=(10, 4))
    raw_selected2.plot_psd(fmax=100, show=False)
    plt.title('PSD After FIR Filtering (Selected 12 Channels)')


# ==========================================================
# ✅ [3] 선택 채널 전처리 및 시각화 (19 Channels) - ✨ 추가된 섹션
# ==========================================================
print("\n" + "="*50)
print("--- [3] 선택 채널 전처리 (19 Channels) ---")

# 10-20 국제 표준 19개 채널 (사용자 요청 목록)
selected_channels_19 = [
    'Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 
    'Fp1', 'Fp2', 'Pz', 'P3', 'P4', 'P7', 'P8', 'O1', 'O2'
]

raw_channels = raw2.ch_names # 채널 이름이 T7, T8로 수정된 상태

# 실제 데이터에 존재하는 채널만 필터링 (오류 방지)
existing_selected_channels_19 = [ch for ch in selected_channels_19 if ch in raw_channels]

# raw19_2 변수에 19개 채널 데이터 복사 및 선택
if not existing_selected_channels_19:
    print(f"\n경고: 요청하신 19개 채널 중 데이터에 존재하는 채널이 없습니다. 전처리를 건너뜁니다.")
    print(f"요청 채널: {selected_channels_19}")
    # 데이터가 없으므로 raw19_2를 생성하지 않습니다.
    raw19_2 = None # 이후 코드에서 NameError 방지를 위해 명시적 할당
else:
    if len(existing_selected_channels_19) < len(selected_channels_19):
        print(f"\n경고: 요청하신 19개 채널 중 {len(selected_channels_19) - len(existing_selected_channels_19)}개가 데이터에 없어 제외되었습니다.")
        print(f"실제 전처리 채널: {existing_selected_channels_19}")

    # 선택한 채널만 복사해서 새 객체 raw19_2로 만들기 
    raw19_2 = raw2.copy().pick_channels(existing_selected_channels_19)
    print(f"raw19_2 객체 생성 완료. 포함 채널: {raw19_2.ch_names}")
    
    # 몽타주 시각화
    if montage_set_successfully:
        try:
            # 선택된 채널의 위치 시각화
            raw19_2.plot_sensors(
                show_names=True,
                kind='topomap', 
                sphere='auto',
                title='Selected 19 Channel Locations (User Defined)'
            )
        except Exception as e:
            print(f"\n[경고] 19개 채널 위치 시각화 실패: {e}")
    else:
        print("\n[알림] 원본 raw2에 몽타주가 설정되지 않아 19개 채널 위치 시각화를 건너뜁니다.")

    # 필터 제거 전 PSD 시각화
    plt.figure(figsize=(10, 4))
    raw19_2.plot_psd(fmax=100, show=False)
    plt.title('PSD Before FIR Filtering (19 Selected Channels)')
    
    # FIR 대역 통과필터: 0.5Hz~ 55Hz
    print("\nraw19_2 FIR 필터링 (0.5Hz ~ 55Hz) 시작...")
    raw19_2.filter(
        l_freq=0.5,
        h_freq=55.0,
        method='fir',
        fir_design='firwin',
        phase='zero',
    )
    print("raw19_2 FIR 필터링 완료.")

    # 필터 제거 후 PSD 시각화
    plt.figure(figsize=(10, 4))
    raw19_2.plot_psd(fmax=100, show=False)
    plt.title('PSD After FIR Filtering (19 Selected Channels)')

# ==========================================================

plt.show() # 모든 시각화 결과(이벤트, 몽타주, PSD 전후)를 한 번에 plot
```

```python
import numpy as np
from scipy.signal import welch
import warnings

# 경고 메시지를 필터링하여 출력하지 않음
warnings.filterwarnings("ignore", category=RuntimeWarning)
warnings.filterwarnings("ignore", category=UserWarning)

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 MNE 데이터 준비 코드에서 이미 정의되어야 합니다.
# X2: (N_epochs_total, 12_ch, N_samp) - 12채널 슬라이싱된 에포크 데이터
# X19_2: (N_epochs_total, 19_ch, N_samp) - 19채널 슬라이싱된 에포크 데이터
# trials2: (N_trials, 3) - MNE 트라이얼 이벤트 배열
# epochs_full2: MNE Epochs 객체 (12채널, tmin=-2.0s 포함)
# epochs_full19_2: MNE Epochs 객체 (19채널, tmin=-2.0s 포함) <--- 19채널용 변수 추가
# sfreq2: 샘플링 주파수
# =========================================================================

# -------------------------------------------------------------------------
# 1. 주파수 밴드 설정
# -------------------------------------------------------------------------
BANDS = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
BASELINE_RANGE = (-2.0, 0.0) # 베이스라인 기간 (t=-2.0s ~ 0.0s)
eps_stability = 1e-12 

# -------------------------------------------------------------------------
# 2. 베이스라인 Log-PSD 및 DE 특징 계산 함수 정의
# -------------------------------------------------------------------------
def compute_baseline_features(X_base, sfreq, bands):
    """
    각 트라이얼 및 채널에 대해 주파수 밴드별 Log-PSD 특징(Mean, Median, Peak)
    및 DE 특징을 계산합니다.
    
    반환: F_mean_base, F_median_base, F_peak_base, F_de_base
          (4개의 (N_trials, N_channels, N_bands) 배열)
    """
    n_trial, n_ch, n_samp_base = X_base.shape
    n_b = len(bands)
    
    # Welch 윈도우 크기 설정 (1초)
    nperseg = int(sfreq * 1.0)
    
    # 결과 배열 초기화: 4가지 특징
    F_mean_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32)
    F_median_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32)
    F_peak_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32)
    F_de_base = np.empty((n_trial, n_ch, n_b), dtype=np.float32) # DE 배열
    
    for i in range(n_trial):
        for ch in range(n_ch):
            channel_data = X_base[i, ch, :]
            
            # Welch PSD 계산
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12  # V^2를 µV^2로 변환
            
            # 주파수 해상도 (DE 계산에 필요)
            df = f[1] - f[0] if len(f) > 1 else 1.0
            
            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                if len(f) > 1 and np.max(f) >= high:
                    idx = np.logical_and(f >= low, f <= high)
                else:
                    idx = np.array([], dtype=bool)
                
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    # 1. Log-PSD 기반 특징 (Mean, Median, Peak)
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_base[i, ch, b_idx] = np.mean(band_log_psd)
                    F_median_base[i, ch, b_idx] = np.median(band_log_psd)
                    F_peak_base[i, ch, b_idx] = np.max(band_log_psd)
                    
                    # 2. 밴드 파워 및 DE 기반 특징 (DE)
                    band_power = np.sum(band_psd) * df
                    
                    if band_power > 0:
                        F_de_base[i, ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability)
                    else:
                        F_de_base[i, ch, b_idx] = 0.0
                    
                else:
                    # 데이터가 없는 경우 0으로 처리
                    F_mean_base[i, ch, b_idx] = 0.0
                    F_median_base[i, ch, b_idx] = 0.0
                    F_peak_base[i, ch, b_idx] = 0.0
                    F_de_base[i, ch, b_idx] = 0.0
                    
    return F_mean_base, F_median_base, F_peak_base, F_de_base

# -------------------------------------------------------------------------
# 3. 트라이얼당 에포크 수 동적 계산 (12채널 데이터 기준)
# -------------------------------------------------------------------------
if 'trials2' in locals() and trials2.shape[0] > 0 and 'X2' in locals() and X2.shape[0] > 0:
    N_EPOCHS_PER_TRIAL2 = X2.shape[0] // trials2.shape[0]
else:
    N_EPOCHS_PER_TRIAL2 = 0
    print("[경고] 트라이얼 또는 에포크 데이터(X2, trials2)가 없어 N_EPOCHS_PER_TRIAL2를 0으로 설정합니다.")
print(f"▶ N_EPOCHS_PER_TRIAL2 (동적 계산, 공유값): {N_EPOCHS_PER_TRIAL2}")


# -------------------------------------------------------------------------
# 4. 베이스라인 데이터 추출 및 특징 계산 (12채널)
# -------------------------------------------------------------------------
print("\n" + "="*50)
print("--- [A] 12채널 베이스라인 특징 계산 ---")

# 12채널 베이스라인 데이터 추출
try:
    epochs_baseline2 = epochs_full2.copy().crop(tmin=BASELINE_RANGE[0], tmax=BASELINE_RANGE[1])
    X_baseline2 = epochs_baseline2.get_data() # (N_trials, 12_ch, N_samp_baseline)
except NameError:
    print("[경고] epochs_full2 변수가 정의되지 않아 12채널 처리를 건너뜁니다.")
    X_baseline2 = np.empty((0,12,int(sfreq2*2)))
except Exception as e:
    print(f"[오류] 12채널 베이스라인 추출 중 오류 발생: {e}")
    X_baseline2 = np.empty((0,12,int(sfreq2*2)))


# 🌟 12채널 특징 계산 실행 (트라이얼당)
print("[OK] 12채널 베이스라인 Log-PSD 및 DE 특징 계산 중...")
F_mean_baseline_trials2, F_median_baseline_trials2, F_peak_baseline_trials2, F_de_baseline_trials2 = \
    compute_baseline_features(X_baseline2, sfreq2, BANDS)

print("[OK] 12채널 베이스라인 Log-PSD 및 DE 특징 계산 완료")
print(f"▶ F_mean_baseline_trials2 (트라이얼 평균): {F_mean_baseline_trials2.shape}")
print(f"▶ F_de_baseline_trials2 (트라이얼 DE): {F_de_baseline_trials2.shape}")

# 🌟 12채널 특징 에포크 매칭
if N_EPOCHS_PER_TRIAL2 > 0:
    F_mean_baseline_matched2 = np.repeat(F_mean_baseline_trials2, N_EPOCHS_PER_TRIAL2, axis=0)
    F_median_baseline_matched2 = np.repeat(F_median_baseline_trials2, N_EPOCHS_PER_TRIAL2, axis=0)
    F_peak_baseline_matched2 = np.repeat(F_peak_baseline_trials2, N_EPOCHS_PER_TRIAL2, axis=0)
    F_de_baseline_matched2 = np.repeat(F_de_baseline_trials2, N_EPOCHS_PER_TRIAL2, axis=0)
    print(f"▶ F_mean_baseline_matched2 (매칭된 평균): {F_mean_baseline_matched2.shape}")
    print(f"▶ F_de_baseline_matched2 (매칭된 DE): {F_de_baseline_matched2.shape}")
else:
    # 매칭 변수 초기화
    n_ch_12 = X_baseline2.shape[1] if X_baseline2.size > 0 else 12
    n_b = len(BANDS)
    F_mean_baseline_matched2 = np.empty((0, n_ch_12, n_b), dtype=np.float32)
    F_median_baseline_matched2 = np.empty((0, n_ch_12, n_b), dtype=np.float32)
    F_peak_baseline_matched2 = np.empty((0, n_ch_12, n_b), dtype=np.float32)
    F_de_baseline_matched2 = np.empty((0, n_ch_12, n_b), dtype=np.float32)


# -------------------------------------------------------------------------
# 5. 베이스라인 데이터 추출 및 특징 계산 (19채널) - ✨ 새로 추가된 섹션
# -------------------------------------------------------------------------
print("\n" + "="*50)
print("--- [B] 19채널 베이스라인 특징 계산 ---")

# 19채널 베이스라인 데이터 추출
try:
    epochs_baseline19_2 = epochs_full19_2.copy().crop(tmin=BASELINE_RANGE[0], tmax=BASELINE_RANGE[1])
    X_baseline19_2 = epochs_baseline19_2.get_data() # (N_trials, 19_ch, N_samp_baseline)
except NameError:
    print("[경고] epochs_full19_2 변수가 정의되지 않아 19채널 처리를 건너뜁니다.")
    X_baseline19_2 = np.empty((0,19,int(sfreq2*2)))
except Exception as e:
    print(f"[오류] 19채널 베이스라인 추출 중 오류 발생: {e}")
    X_baseline19_2 = np.empty((0,19,int(sfreq2*2)))


# 🌟 19채널 특징 계산 실행 (트라이얼당)
print("[OK] 19채널 베이스라인 Log-PSD 및 DE 특징 계산 중...")
# 변수명: F_[특징]_baseline_trials19_2
F_mean_baseline_trials19_2, F_median_baseline_trials19_2, F_peak_baseline_trials19_2, F_de_baseline_trials19_2 = \
    compute_baseline_features(X_baseline19_2, sfreq2, BANDS)

print("[OK] 19채널 베이스라인 Log-PSD 및 DE 특징 계산 완료")
print(f"▶ F_mean_baseline_trials19_2 (트라이얼 평균): {F_mean_baseline_trials19_2.shape}")
print(f"▶ F_de_baseline_trials19_2 (트라이얼 DE): {F_de_baseline_trials19_2.shape}")

# 🌟 19채널 특징 에포크 매칭
if N_EPOCHS_PER_TRIAL2 > 0:
    # 변수명: F_[특징]_baseline_matched19_2
    F_mean_baseline_matched19_2 = np.repeat(F_mean_baseline_trials19_2, N_EPOCHS_PER_TRIAL2, axis=0)
    F_median_baseline_matched19_2 = np.repeat(F_median_baseline_trials19_2, N_EPOCHS_PER_TRIAL2, axis=0)
    F_peak_baseline_matched19_2 = np.repeat(F_peak_baseline_trials19_2, N_EPOCHS_PER_TRIAL2, axis=0)
    F_de_baseline_matched19_2 = np.repeat(F_de_baseline_trials19_2, N_EPOCHS_PER_TRIAL2, axis=0)

    print("[OK] 19채널 베이스라인 Log-PSD 및 DE 특징 에포크 매칭 완료")
    print(f"▶ F_mean_baseline_matched19_2 (매칭된 평균): {F_mean_baseline_matched19_2.shape}")
    print(f"▶ F_de_baseline_matched19_2 (매칭된 DE): {F_de_baseline_matched19_2.shape}")
else:
    # 매칭 변수 초기화
    n_ch_19 = X_baseline19_2.shape[1] if X_baseline19_2.size > 0 else 19
    n_b = len(BANDS)
    F_mean_baseline_matched19_2 = np.empty((0, n_ch_19, n_b), dtype=np.float32)
    F_median_baseline_matched19_2 = np.empty((0, n_ch_19, n_b), dtype=np.float32)
    F_peak_baseline_matched19_2 = np.empty((0, n_ch_19, n_b), dtype=np.float32)
    F_de_baseline_matched19_2 = np.empty((0, n_ch_19, n_b), dtype=np.float32)

print("="*50)
```

```python
import numpy as np

def print_group_label_details(groups2, y2):
    """
    각 그룹(트라이얼)의 라벨 분포를 상세히 출력합니다.
    """
    print("\n" + "="*50)
    print("      groups2 & y2 데이터 그룹화 상세 정보")
    print("="*50)

    if groups2.shape != y2.shape:
        print(f"❌ 오류: groups2와 y2의 차원이 다릅니다. groups2: {groups2.shape}, y2: {y2.shape}")
        return

    unique_groups = np.unique(groups2)
    
    # 각 그룹의 상세 정보를 출력
    for g2 in unique_groups:
        # 현재 그룹에 속한 에포크들의 인덱스 추출
        group_indices = np.where(groups2 == g2)[0]
        
        # 해당 인덱스들의 라벨 값 추출
        group_labels = y2[group_indices]
        
        # 라벨이 모두 같은지 확인
        unique_labels_in_group = np.unique(group_labels)
        
        status = "✅ 일치" if len(unique_labels_in_group) == 1 else "❌ 불일치"
        
        print(f"▶ 그룹 ID: {g2} | 상태: {status}")
        print(f"  - 속한 에포크 수: {len(group_indices)}")
        print(f"  - 발견된 라벨: {unique_labels_in_group}")
        
        # 라벨이 불일치하는 경우 상세 분포 출력
        if status == "❌ 불일치":
            label_counts = {label: np.sum(group_labels == label) for label in unique_labels_in_group}
            print(f"  - 라벨 분포: {label_counts}")
        print("-" * 50)

    print("\n▶ 요약:")
    print(f"  - 총 그룹(트라이얼) 수: {len(unique_groups)}")
    mismatched_groups = [g2 for g2 in unique_groups if len(np.unique(y2[np.where(groups2 == g2)[0]])) > 1]
    print(f"  - 라벨 불일치 그룹 수: {len(mismatched_groups)}")
    print("="*50 + "\n")

# --------------------------------------------------------------------
# 사용 예시:
# 이 코드를 기존 스크립트의 groups2와 y2 변수가 생성된 후에 추가하여 호출하세요.
# 예시:
# F_3d2 = compute_psd_log10_epochs_3d2(X2, y2, groups2, sfreq2, BANDS2)
print_group_label_details(groups2, y2)
```

```python
import numpy as np
from scipy.signal import welch
import warnings

# 경고 메시지를 필터링하여 출력하지 않음
warnings.filterwarnings("ignore", category=RuntimeWarning)
warnings.filterwarnings("ignore", category=UserWarning)

# =========================================================================
# ⚠️ 전제 조건: 아래 변수들이 이미 정의되어야 합니다.
# X2: (n_epochs, 12_ch, n_samp) - 12채널 슬라이싱된 이벤트 기간 EEG 데이터
# X19_2: (n_epochs, 19_ch, n_samp) - 19채널 슬라이싱된 이벤트 기간 EEG 데이터
# y2, groups2, sfreq2: 레이블, 그룹, 샘플링 주파수
#
# [이전 베이스라인 코드에서 생성된 변수들]
# F_[특징]_baseline_matched2: (n_epochs, 12, n_bands)
# F_[특징]_baseline_matched19_2: (n_epochs, 19, n_bands) <--- 19채널 베이스라인
# =========================================================================

# -------------------------------------------------------------------------
# 1. 주파수 밴드 및 설정
# -------------------------------------------------------------------------
BANDS2 = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13),
    "beta": (13, 30), "gamma": (30, 50)
}
band_keys2 = list(BANDS2.keys())
eps_stability = 1e-12

# -------------------------------------------------------------------------
# 2. 이벤트 기간 특징 계산 함수 (Mean, Median, Peak, DE) - 함수는 재사용
# -------------------------------------------------------------------------
def compute_event_features_3d(X, sfreq, bands):
    """
    이벤트 기간(X)에 대해 4가지 특징(Log-PSD Mean, Median, Peak 및 DE)을 계산합니다.
    """
    n_ep, n_ch, n_samp = X.shape
    n_b = len(bands)
    
    # Welch 윈도우 크기 설정 (1초)
    nperseg = int(sfreq)
    
    # 결과 배열 초기화: 4가지 특징
    F_mean_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_median_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_peak_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)
    F_de_event = np.empty((n_ep, n_ch, n_b), dtype=np.float32)

    for i in range(n_ep):
        for ch in range(n_ch):
            channel_data = X[i, ch, :]
            
            # Welch PSD 계산
            f, Pxx = welch(channel_data, fs=sfreq, nperseg=nperseg)
            Pxx_uv = Pxx * 1e12  # V^2를 µV^2로 변환
            
            # 주파수 해상도 (DE 계산에 필요)
            df = f[1] - f[0] if len(f) > 1 else 1.0
            
            for b_idx, (band, (low, high)) in enumerate(bands.items()):
                if len(f) > 1 and np.max(f) >= high:
                    idx = np.logical_and(f >= low, f <= high)
                else:
                    idx = np.array([], dtype=bool)
                
                band_psd = Pxx_uv[idx]
                
                if len(band_psd) > 0:
                    # 1. Log-PSD 기반 특징 (Mean, Median, Peak)
                    band_log_psd = np.log10(band_psd + eps_stability)
                    F_mean_event[i, ch, b_idx] = np.mean(band_log_psd)
                    F_median_event[i, ch, b_idx] = np.median(band_log_psd)
                    F_peak_event[i, ch, b_idx] = np.max(band_log_psd)
                    
                    # 2. 밴드 파워 및 DE 기반 특징 (DE)
                    band_power = np.sum(band_psd) * df
                    
                    if band_power > 0:
                        F_de_event[i, ch, b_idx] = 0.5 * np.log(2 * np.pi * np.e * band_power + eps_stability)
                    else:
                        F_de_event[i, ch, b_idx] = 0.0
                    
                else:
                    # 데이터가 없는 경우 0으로 처리
                    F_mean_event[i, ch, b_idx] = 0.0
                    F_median_event[i, ch, b_idx] = 0.0
                    F_peak_event[i, ch, b_idx] = 0.0
                    F_de_event[i, ch, b_idx] = 0.0
                    
    return F_mean_event, F_median_event, F_peak_event, F_de_event


# =========================================================================
# 3. [A] 12채널 특징 추출 및 정규화
# =========================================================================
print("="*60)
print("--- [A] 12채널 특징 추출 및 정규화 ---")

# 3-1. X2에서 이벤트 특징 추출
try:
    print("[OK] 12채널 이벤트 기간 Log-PSD 및 DE 특징 계산 중...")
    F_mean_event2, F_median_event2, F_peak_event2, F_de_event2 = \
        compute_event_features_3d(X2, sfreq2, BANDS2)
    print("[OK] 12채널 이벤트 기간 특징 계산 완료.")

    # 3-2. 로그 비율 정규화 (Log Ratio Normalization)
    print("\n[OK] 12채널 Log Ratio 정규화 (차분) 적용 중...")
    F_mean_3d2 = F_mean_event2 - F_mean_baseline_matched2
    F_median_3d2 = F_median_event2 - F_median_baseline_matched2
    F_peak_3d2 = F_peak_event2 - F_peak_baseline_matched2
    F_de_3d2 = F_de_event2 - F_de_baseline_matched2
    print("[OK] 12채널 Log Ratio 정규화 완료.")

    # 3-3. 비대칭 특징 (DASM, RASM) 계산
    CHANNELS2 = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2']
    HEM_PAIRS2 = [ ('F3', 'F4'), ('F7', 'F8'), ('C3', 'C4'), ('T7', 'T8'), ('Fp1', 'Fp2') ]
    pair_names2 = [f"{L}-{R}" for L, R in HEM_PAIRS2]  
    ch_to_idx2 = {ch_name: i for i, ch_name in enumerate(CHANNELS2)}
    n_pairs2 = len(HEM_PAIRS2)

    n_ep2, _, n_b2 = F_de_3d2.shape
    F_dasm_3d_lr2 = np.empty((n_ep2, n_pairs2, n_b2), dtype=np.float32)
    F_rasm_3d_lr2 = np.empty((n_ep2, n_pairs2, n_b2), dtype=np.float32)

    for i, (left_ch, right_ch) in enumerate(HEM_PAIRS2):
        left_idx = ch_to_idx2[left_ch]
        right_idx = ch_to_idx2[right_ch]
        de_norm_left = F_de_3d2[:, left_idx, :]
        de_norm_right = F_de_3d2[:, right_idx, :]
        
        F_dasm_3d_lr2[:, i, :] = de_norm_left - de_norm_right
        F_rasm_3d_lr2[:, i, :] = de_norm_left / (de_norm_right + eps_stability)

    F_dasm_3d2 = F_dasm_3d_lr2.copy()
    F_rasm_3d2 = F_rasm_3d_lr2.copy()
    print("[OK] 12채널 비대칭 특징 (DASM, RASM) 생성 완료.")

except NameError as e:
    print(f"[경고] 12채널 처리 중 변수 오류 발생 ({e}). 12채널 처리를 건너뛰고 빈 배열을 생성합니다.")
    n_ch_12 = 12
    n_b = len(BANDS2)
    F_mean_3d2 = F_median_3d2 = F_peak_3d2 = F_de_3d2 = np.empty((0, n_ch_12, n_b), np.float32)
    F_dasm_3d2 = F_rasm_3d2 = np.empty((0, 5, n_b), np.float32) # 5쌍 기준
    pair_names2 = []


# =========================================================================
# 4. [B] 19채널 특징 추출 및 정규화 - ✨ 새로 추가된 섹션
# =========================================================================
print("="*60)
print("--- [B] 19채널 특징 추출 및 정규화 ---")

try:
    if 'X19_2' not in locals():
        raise NameError("X19_2")
        
    print("[OK] 19채널 이벤트 기간 Log-PSD 및 DE 특징 계산 중...")
    # 변수명: F_[특징]_event19_2
    F_mean_event19_2, F_median_event19_2, F_peak_event19_2, F_de_event19_2 = \
        compute_event_features_3d(X19_2, sfreq2, BANDS2)
    print("[OK] 19채널 이벤트 기간 특징 계산 완료.")

    # 4-1. 로그 비율 정규화 (19채널)
    print("\n[OK] 19채널 Log Ratio 정규화 (차분) 적용 중...")
    # 변수명: F_[특징]_3d19_2
    F_mean_3d19_2 = F_mean_event19_2 - F_mean_baseline_matched19_2
    F_median_3d19_2 = F_median_event19_2 - F_median_baseline_matched19_2
    F_peak_3d19_2 = F_peak_event19_2 - F_peak_baseline_matched19_2
    F_de_3d19_2 = F_de_event19_2 - F_de_baseline_matched19_2
    print("[OK] 19채널 Log Ratio 정규화 완료.")

    # 4-2. 비대칭 특징 (DASM, RASM) 계산 - 정규화된 DE(F_de_3d19_2) 기반
    
    # 19채널 목록 및 쌍 정의
    CHANNELS19_2 = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2', 'Pz', 'P3', 'P4', 'P7', 'P8', 'O1', 'O2']
    # 19채널 중에서 반구 쌍만 추출 (8쌍)
    HEM_PAIRS19_2 = [ ('F3', 'F4'), ('F7', 'F8'), ('C3', 'C4'), ('T7', 'T8'), ('Fp1', 'Fp2'), ('P3', 'P4'), ('P7', 'P8'), ('O1', 'O2') ]
    pair_names19_2 = [f"{L}-{R}" for L, R in HEM_PAIRS19_2]  
    ch_to_idx19_2 = {ch_name: i for i, ch_name in enumerate(CHANNELS19_2)}
    n_pairs19_2 = len(HEM_PAIRS19_2)

    # 특징 배열 초기화
    n_ep19_2, _, n_b19_2 = F_de_3d19_2.shape
    F_dasm_3d_lr19_2 = np.empty((n_ep19_2, n_pairs19_2, n_b19_2), dtype=np.float32)
    F_rasm_3d_lr19_2 = np.empty((n_ep19_2, n_pairs19_2, n_b19_2), dtype=np.float32)

    # 각 채널 쌍에 대해 DASM, RASM 계산
    for i, (left_ch, right_ch) in enumerate(HEM_PAIRS19_2):
        left_idx = ch_to_idx19_2[left_ch]  
        right_idx = ch_to_idx19_2[right_ch]
        
        # 정규화된 DE 값 추출
        de_norm_left = F_de_3d19_2[:, left_idx, :]
        de_norm_right = F_de_3d19_2[:, right_idx, :]
        
        # DASM 계산: Left - Right
        F_dasm_3d_lr19_2[:, i, :] = de_norm_left - de_norm_right
        
        # RASM 계산: Left / Right
        F_rasm_3d_lr19_2[:, i, :] = de_norm_left / (de_norm_right + eps_stability)

    # 최종 변수명: F_[특징]_3d19_2
    F_dasm_3d19_2 = F_dasm_3d_lr19_2.copy()
    F_rasm_3d19_2 = F_rasm_3d_lr19_2.copy()
    print("[OK] 19채널 비대칭 특징 (DASM, RASM) 생성 완료.")
    
except NameError as e:
    print(f"[경고] 19채널 처리 중 변수 오류 발생 ({e}). 19채널 처리를 건너뛰고 빈 배열을 생성합니다.")
    n_ch_19 = 19
    n_b = len(BANDS2)
    F_mean_3d19_2 = F_median_3d19_2 = F_peak_3d19_2 = F_de_3d19_2 = np.empty((0, n_ch_19, n_b), np.float32)
    F_dasm_3d19_2 = F_rasm_3d19_2 = np.empty((0, 8, n_b), np.float32) # 8쌍 기준
    pair_names19_2 = []
except Exception as e:
    print(f"[오류] 19채널 특징 계산 중 예측하지 못한 오류 발생: {e}")
    n_ch_19 = 19
    n_b = len(BANDS2)
    F_mean_3d19_2 = F_median_3d19_2 = F_peak_3d19_2 = F_de_3d19_2 = np.empty((0, n_ch_19, n_b), np.float32)
    F_dasm_3d19_2 = F_rasm_3d19_2 = np.empty((0, 8, n_b), np.float32)


# -------------------------------------------------------------------------
# 5. 최종 결과 확인
# -------------------------------------------------------------------------
print("="*60)
print("--- 최종 결과 비교 ---")

### 12채널 특징 확인
print("\n[12채널 특징 (suffix: _3d2)]")
print(f"  - F_mean_3d2 (Log-PSD Mean 차분): {F_mean_3d2.shape}")
print(f"  - F_de_3d2 (DE 차분): {F_de_3d2.shape}")
print(f"  - F_dasm_3d2 (DASM): {F_dasm_3d2.shape}")
print(f"  - F_rasm_3d2 (RASM): {F_rasm_3d2.shape}")
if pair_names2:
    print(f"  - DASM/RASM 채널 쌍 순서: {pair_names2}")
else:
    print("  - DASM/RASM 채널 쌍 정보 없음.")

### 19채널 특징 확인
print("\n[19채널 특징 (suffix: _3d19_2)]")
print(f"  - F_mean_3d19_2 (Log-PSD Mean 차분): {F_mean_3d19_2.shape}")
print(f"  - F_de_3d19_2 (DE 차분): {F_de_3d19_2.shape}")
print(f"  - F_dasm_3d19_2 (DASM): {F_dasm_3d19_2.shape}")
print(f"  - F_rasm_3d19_2 (RASM): {F_rasm_3d19_2.shape}")
if pair_names19_2:
    print(f"  - DASM/RASM 채널 쌍 순서: {pair_names19_2}")
else:
    print("  - DASM/RASM 채널 쌍 정보 없음.")
print("="*60)
```

```python
#메소드2
import numpy as np
import time
import warnings
import xgboost as xgb
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
from scipy.stats import mode
from itertools import product
from tabulate import tabulate

# ⚠️ 주의: 이 코드를 실행하려면 F_mean_3d 등 모든 특징 변수, y, groups, y2, groups2 변수가 전역적으로 정의되어 있어야 합니다.

CHANNELS = ['Fz', 'F3', 'F4', 'F7', 'F8', 'Cz', 'C3', 'C4', 'T7', 'T8', 'Fp1', 'Fp2'] # 12개 채널 정의

n_trials_vis = 30
n_epochs_per_trial = 7
n_trials_olf = 20
all_results = []

# --- 채널 및 밴드 정의 (U+00A0 제거 후 수정) ---
BANDS2 = {
    "delta": (1, 4), "theta": (4, 8), "alpha": (8, 13), "beta": (13, 30), "gamma": (30, 50)
}
BAND_NAMES = list(BANDS2.keys())
BAND_CONDITIONS = BAND_NAMES + ['all']


# ------------------------------------------------------------------------------
# ⭐ 3. 단일 채널 특징 (48쌍) 딕셔너리 자동 통합 ⭐
# ------------------------------------------------------------------------------
print("🧠 단일 채널 특징 48쌍을 VIS_FEATURES_ALL과 OLF_FEATURES_ALL에 통합합니다.")

FEATURE_TYPES_SINGLE_CH = ['mean', 'median', 'peak', 'de']
channels_to_integrate = CHANNELS # ['Fz', 'F3', ..., 'Fp2']

count_vis = 0
count_olf = 0

# VIS_FEATURES_ALL과 OLF_FEATURES_ALL 딕셔너리를 초기화합니다.
try:
    _vis_temp_copy = VIS_FEATURES_ALL.copy()
    _olf_temp_copy = OLF_FEATURES_ALL.copy()
except NameError:
    _vis_temp_copy = {}
    _olf_temp_copy = {}
    
VIS_FEATURES_ALL = {}
OLF_FEATURES_ALL = {}
VIS_FEATURES_ALL.update(_vis_temp_copy)
OLF_FEATURES_ALL.update(_olf_temp_copy)


for ch in channels_to_integrate:
    for feat_type in FEATURE_TYPES_SINGLE_CH:
        vis_name = f'F_{feat_type}_3d_{ch}'
        olf_name = f'F_{feat_type}_3d2_{ch}'
        
        # 1. 시각 특징 (훈련 데이터) 통합
        if vis_name in globals():
            VIS_FEATURES_ALL[vis_name] = globals()[vis_name]
            count_vis += 1
        
        # 2. 후각 특징 (테스트 데이터) 통합
        if olf_name in globals():
            OLF_FEATURES_ALL[olf_name] = globals()[olf_name]
            count_olf += 1

print(f"✅ VIS_FEATURES_ALL에 {count_vis}개의 단일 채널 특징 통합 완료.")
print(f"✅ OLF_FEATURES_ALL에 {count_olf}개의 단일 채널 특징 통합 완료.")
print(f"--- 통합 후 특징 쌍 개수: {len(VIS_FEATURES_ALL.keys())} / {len(OLF_FEATURES_ALL.keys())} ---")


# --- 특징 조합 구성 로직 (단일 채널 포함하도록 수정) ---
FEATURE_TYPES = ['mean', 'median', 'peak', 'de', 'dasm', 'rasm']

all_feature_combinations = {}
total_combinations_count = 0

# 1. 단일 채널 특징 (48개) 통합
for ch_name in CHANNELS:
    for feat_type in ['mean', 'median', 'peak', 'de']: # 단일 채널은 4가지 특징만 사용
        vis_name = f'F_{feat_type}_3d_{ch_name}'
        olf_name = f'F_{feat_type}_3d2_{ch_name}'
        
        if vis_name in VIS_FEATURES_ALL and olf_name in OLF_FEATURES_ALL:
            all_feature_combinations[vis_name] = (VIS_FEATURES_ALL[vis_name], OLF_FEATURES_ALL[olf_name])
            total_combinations_count += 1

# 2. 그룹 특징 통합 (32개 -> 38개)
CHANNEL_CONFIGS_GROUP = [
    ('', ''), ('10rm', '10rm'), ('8rm', '8rm'),
    ('2area', '2area'), ('1area', '1area'), ('3area', '3area'),
    ('19', '19') # 💡 수정: F_..._3d19 특징을 위한 '19' 접미사 추가
]

for suffix, identifier in CHANNEL_CONFIGS_GROUP:
    if identifier in ['2area', '1area']:
        feat_types = [t for t in FEATURE_TYPES if t not in ['dasm', 'rasm']]
    else:
        feat_types = FEATURE_TYPES
    
    base_suffix = '' if suffix == '' else f'_{suffix}'
    
    for feat_type in feat_types:
        
        # 💡 수정 1/2: 19채널 시각 변수가 F_mean_3d19 형태로 존재함을 반영
        if identifier == '19':
            vis_name = f'F_{feat_type}_3d{identifier}' # 예: F_mean_3d19
        else:
            vis_name = f'F_{feat_type}_3d{base_suffix}'
        
        # 🌟 핵심 수정 2/2: 19채널 그룹의 olf_name이 F_mean_3d19_2 형태로 존재함을 반영 🌟
        if identifier == '19':
            # F_mean_3d19_2, F_dasm_3d19_2 등을 찾도록 변수명 직접 구성
            olf_name = f'F_{feat_type}_3d{identifier}_2'
        else:
            # 나머지 그룹은 기존의 'vis_name + 2' 방식을 유지 
            olf_name = vis_name + '2'
        
        # 💡 globals()에서 변수가 존재하는지 확인하고 통합
        if vis_name in globals() and olf_name in globals():
            VIS_FEATURES_ALL[vis_name] = globals()[vis_name]
            OLF_FEATURES_ALL[olf_name] = globals()[olf_name]
            all_feature_combinations[vis_name] = (globals()[vis_name], globals()[olf_name])
            total_combinations_count += 1

# 🚨 최종 특징 조합 개수 확인 (80개 -> 86개)
expected_count = 86
if len(all_feature_combinations) != expected_count:
    print(f"⚠️ 경고: 생성된 특징 쌍 개수가 예상치({expected_count}개)와 다릅니다. 현재: {len(all_feature_combinations)}개")
    print("      (F_..._3d19 특징 변수가 전역적으로 정의되어 있는지 확인하세요.)")
# ======================================================================
# ✅ 정규화 및 정확도 계산 함수 정의 (유지)
# ======================================================================

def normalize_norm2(F_3d_data, groups_data, band_index=None):
    """그룹(피험자)-채널/영역-밴드 단위 정규화. 특정 밴드만 사용 가능."""
    if F_3d_data.size == 0 or F_3d_data.ndim != 3: return np.empty_like(F_3d_data, dtype=np.float32)
    
    if band_index is None:
        F_slice = F_3d_data
    else:
        F_slice = F_3d_data[..., band_index:band_index+1]
                
    F_standardized_temp = np.zeros_like(F_slice, dtype=np.float32)
    unique_groups = np.unique(groups_data)
    
    n_channels = F_slice.shape[1]
    n_bands_slice = F_slice.shape[2]
            
    for group_id in unique_groups:
        group_indices = np.where(groups_data == group_id)[0]
        if group_indices.size == 0: continue
        for channel_idx in range(n_channels):
            for band_idx in range(n_bands_slice):
                group_channel_band_data = F_slice[group_indices, channel_idx, band_idx]
                mu = np.mean(group_channel_band_data)
                sigma = np.std(group_channel_band_data) + 1e-12
                F_standardized_temp[group_indices, channel_idx, band_idx] = (group_channel_band_data - mu) / sigma
    
    # 2D 형태로 평탄화하여 반환
    return F_standardized_temp.reshape(F_standardized_temp.shape[0], -1)


def normalize_norm3(F_3d_data, groups_data, band_index=None):
    """그룹(피험자)-전체 데이터 통합 정규화. 특정 밴드만 사용 가능."""
    if F_3d_data.size == 0 or F_3d_data.ndim != 3: return np.empty_like(F_3d_data, dtype=np.float32)
    
    if band_index is None:
        F_slice = F_3d_data
    else:
        F_slice = F_3d_data[..., band_index:band_index+1]
    
    F_standardized_temp = np.zeros_like(F_slice, dtype=np.float32)
    unique_groups = np.unique(groups_data)
    
    for group_id in unique_groups:
        group_indices = np.where(groups_data == group_id)[0]
        if group_indices.size == 0: continue
        group_all_data = F_slice[group_indices, ...]
        mu = np.mean(group_all_data.flatten())
        sigma = np.std(group_all_data.flatten()) + 1e-12
        F_standardized_temp[group_indices, ...] = (group_all_data - mu) / sigma
        
    # 2D 형태로 평탄화하여 반환
    return F_standardized_temp.reshape(F_standardized_temp.shape[0], -1)


def calculate_accuracy(y_pred_epoch, y_true_epoch, group_trial_ids, y_true_trial):
    if len(y_true_epoch) == 0: return 0.0, 0.0, None

    unique_trials = np.unique(group_trial_ids)
    
    if len(unique_trials) != len(y_true_trial) or len(y_pred_epoch) != len(y_true_epoch):
        # 데이터 불일치 또는 길이가 다를 경우 trial acc는 0으로 반환
        return 0.0, accuracy_score(y_true_epoch, y_pred_epoch), confusion_matrix(y_true_epoch, y_pred_epoch, labels=[0, 1])

    y_pred_trial_agg = []
    for trial_id in unique_trials:
        epoch_indices = np.where(group_trial_ids == trial_id)[0]
        epoch_preds_for_this_trial = y_pred_epoch[epoch_indices]
        if len(epoch_preds_for_this_trial) == 0: continue
        # Trial-level prediction by majority vote
        trial_pred, _ = mode(epoch_preds_for_this_trial, keepdims=False)
        y_pred_trial_agg.append(trial_pred)
    
    if len(y_pred_trial_agg) != len(y_true_trial):
        # Trial 예측 개수가 실제 Trial 개수와 다를 경우
        return 0.0, accuracy_score(y_true_epoch, y_pred_epoch), confusion_matrix(y_true_epoch, y_pred_epoch, labels=[0, 1])


    trial_accuracy = accuracy_score(y_true_trial, y_pred_trial_agg)
    epoch_accuracy = accuracy_score(y_true_epoch, y_pred_epoch)
    cm_epoch = confusion_matrix(y_true_epoch, y_pred_epoch, labels=[0, 1])
    return trial_accuracy, epoch_accuracy, cm_epoch

# ======================================================================
# ✅ 모델 및 하이퍼파라미터 설정 (유지)
# ======================================================================
models_dict = {
    'Logistic Regression': LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42, C=0.1),
    'KNN': KNeighborsClassifier(n_neighbors=5, weights='uniform'),
    'Linear SVM': LinearSVC(max_iter=20000, class_weight='balanced', random_state=42, C=0.1, dual=True),
    'Random Forest': RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42, max_depth=3, min_samples_leaf=20),
    'XGBoost': xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42, max_depth=2, gamma=1.0, min_child_weight=10)
}
model_names = list(models_dict.keys())
all_model_names = model_names + ['Ensemble (Majority Vote)']
norm_types = ['None', 'norm2', 'norm3']
# ======================================================================


# --- 데이터 분할 설정 (유지) ---
# NOTE: y, groups, y2, groups2 변수가 전역적으로 정의되어 있다고 가정
try:
    # 훈련 데이터 (시각)
    groups_trial_id_vis = np.repeat(np.arange(n_trials_vis), n_epochs_per_trial)[:len(y)]
    y_train_epoch = y
    groups_train_subject = groups
    groups_train_trial_id = groups_trial_id_vis
    y_train_trial = y[::n_epochs_per_trial][:n_trials_vis]

    # 테스트 데이터 (후각)
    groups_trial_id_olf = np.repeat(np.arange(n_trials_olf), n_epochs_per_trial)[:len(y2)]
    y_test_epoch = y2
    groups_test_subject = groups2
    groups_test_trial_id = groups_trial_id_olf
    y_test_trial = y2[::n_epochs_per_trial][:n_trials_olf]

    # 데이터 길이 확인
    if len(y) == 0 or len(y2) == 0:
        raise ValueError("전역 변수 y 또는 y2의 길이가 0입니다. 특징 변수 및 레이블이 로드되었는지 확인하세요.")
        
except NameError:
    print("❌ 에러: y, groups, y2, groups2 변수가 전역적으로 정의되어 있지 않아 분석을 시작할 수 없습니다.")
    # 임시 데이터 정의 (분석은 안되지만 코드 구조는 확인 가능)
    # 실제 환경에서는 이 부분이 실행되면 안 됩니다.
    # 이 부분을 실행하면 뒤에서 len(y) 등에 접근할 때 오류가 발생하지 않도록 임시 데이터를 정의합니다.
    y = y2 = groups = groups2 = np.array([0, 1, 0, 1, 0, 1, 0] * 3)
    groups_trial_id_vis = np.repeat(np.arange(n_trials_vis), n_epochs_per_trial)[:len(y)]
    y_train_epoch = y
    groups_train_subject = groups
    groups_train_trial_id = groups_trial_id_vis
    y_train_trial = y[::n_epochs_per_trial][:n_trials_vis]

    groups_trial_id_olf = np.repeat(np.arange(n_trials_olf), n_epochs_per_trial)[:len(y2)]
    y_test_epoch = y2
    groups_test_subject = groups2
    groups_test_trial_id = groups_trial_id_olf
    y_test_trial = y2[::n_epochs_per_trial][:n_trials_olf]
    
except ValueError as e:
    print(f"❌ 데이터 로드 에러: {e}")
    
# ======================================================================
# 🚀 자동화 루프 실행
# ======================================================================

all_results = []
start_time_total = time.time()

# 🚨 총 경우의 수 계산: 특징 86 * 밴드 6 * 정규화 3 * 모델 6 = 9,396가지
total_cases = len(all_feature_combinations) * len(BAND_CONDITIONS) * len(norm_types) * len(all_model_names)
case_count = 0
print(f"특징 쌍 개수: {len(all_feature_combinations)}개")
print(f"밴드 조건 개수: {len(BAND_CONDITIONS)}개")
print(f"모델 개수: {len(all_model_names)}개")
print(f"정규화 개수: {len(norm_types)}개")

print(f"\n================================================================")
print(f"      🧠 총 {total_cases}가지 Cross-Modal 자동화 분석 시작 (밴드 포함)")
print(f"      (Train: Vis ({len(y)} epochs), Test: Olf ({len(y2)} epochs))")
print(f"================================================================")


# 밴드 루프 추가
for feat_name, (raw_3d_train_all_bands, raw_3d_test_all_bands) in all_feature_combinations.items():
    
    for band_condition in BAND_CONDITIONS:
        
        # 밴드 인덱스 결정
        band_idx = BAND_NAMES.index(band_condition) if band_condition != 'all' else None
        
        if band_condition == 'all':
            raw_3d_train = raw_3d_train_all_bands
            raw_3d_test = raw_3d_test_all_bands
        else:
            # 단일 밴드만 사용 (차원 감소: (N, C, 5) -> (N, C, 1))
            raw_3d_train = raw_3d_train_all_bands[..., band_idx:band_idx+1]
            raw_3d_test = raw_3d_test_all_bands[..., band_idx:band_idx+1]

        
        for norm_type in norm_types:
            
            # 1. 훈련 데이터 정규화 적용
            norm_band_idx = 0 if band_condition != 'all' else None
            
            if norm_type == 'None':
                X_train_epoch = raw_3d_train.reshape(raw_3d_train.shape[0], -1)
            elif norm_type == 'norm2':
                # raw_3d_train을 전달하고, normalize_norm2 내부에서 band_index를 사용해 슬라이싱/정규화
                X_train_epoch = normalize_norm2(raw_3d_train, groups_train_subject, band_index=norm_band_idx)
            elif norm_type == 'norm3':
                X_train_epoch = normalize_norm3(raw_3d_train, groups_train_subject, band_index=norm_band_idx)
            else: continue

            # 2. 테스트 데이터 정규화 적용 (수정된 부분)
            if norm_type == 'None':
                # 정규화 없음: 단순히 평탄화
                X_test_epoch = raw_3d_test.reshape(raw_3d_test.shape[0], -1)
            elif norm_type == 'norm2':
                # 테스트 데이터에 norm2 적용 (테스트 데이터의 그룹별 통계 사용)
                X_test_epoch = normalize_norm2(raw_3d_test, groups_test_subject, band_index=norm_band_idx)
            elif norm_type == 'norm3':
                # 테스트 데이터에 norm3 적용 (테스트 데이터의 그룹별 통계 사용)
                X_test_epoch = normalize_norm3(raw_3d_test, groups_test_subject, band_index=norm_band_idx)
            else: continue

            # 3. 모델 훈련 및 예측
            train_preds_list = []
            test_preds_list = []

            for model_name, model in models_dict.items():
                case_count += 1
                
                # 모델 파라미터를 유지하면서 새 인스턴스 생성
                current_model = type(model)().set_params(**model.get_params())
                
                with warnings.catch_warnings():
                    warnings.filterwarnings("ignore")
                    current_model.fit(X_train_epoch, y_train_epoch)
                
                y_pred_train_epoch = current_model.predict(X_train_epoch)
                y_pred_test_epoch = current_model.predict(X_test_epoch)
                
                train_preds_list.append(y_pred_train_epoch)
                test_preds_list.append(y_pred_test_epoch)
                
                # 정확도 및 CM 계산
                train_acc_trial, train_acc_epoch, cm_train_epoch = calculate_accuracy(
                    y_pred_train_epoch, y_train_epoch, groups_train_trial_id, y_train_trial
                )
                test_acc_trial, test_acc_epoch, cm_test_epoch = calculate_accuracy(
                    y_pred_test_epoch, y_test_epoch, groups_test_trial_id, y_test_trial
                )
                
                all_results.append({
                    '특징': feat_name, '밴드': band_condition, '정규화': norm_type, '모델': model_name,
                    '학습 정확도 (Trial)': train_acc_trial, '학습 정확도 (Epoch)': train_acc_epoch,
                    '테스트 정확도 (Trial)': test_acc_trial, '테스트 정확도 (Epoch)': test_acc_epoch,
                    '학습 혼동 행렬 (Epoch)': cm_train_epoch.tolist() if cm_train_epoch is not None else None,
                    '테스트 혼동 행렬 (Epoch)': cm_test_epoch.tolist() if cm_test_epoch is not None else None,
                })

            # 4. 앙상블 (Majority Vote)
            case_count += 1
            preds_stack_train_t = np.array(train_preds_list).T
            preds_stack_test_t = np.array(test_preds_list).T
            
            if preds_stack_train_t.size > 0:
                majority_vote_epoch_train = mode(preds_stack_train_t, axis=1, keepdims=False)[0].astype(int)
            else:
                majority_vote_epoch_train = np.array([])

            if preds_stack_test_t.size > 0:
                majority_vote_epoch_test = mode(preds_stack_test_t, axis=1, keepdims=False)[0].astype(int)
            else:
                majority_vote_epoch_test = np.array([])


            train_acc_trial, train_acc_epoch, cm_train_epoch = calculate_accuracy(
                majority_vote_epoch_train, y_train_epoch, groups_train_trial_id, y_train_trial
            )
            test_acc_trial, test_acc_epoch, cm_test_epoch = calculate_accuracy(
                majority_vote_epoch_test, y_test_epoch, groups_test_trial_id, y_test_trial
            )

            all_results.append({
                '특징': feat_name, '밴드': band_condition, '정규화': norm_type, '모델': 'Ensemble (Majority Vote)',
                '학습 정확도 (Trial)': train_acc_trial, '학습 정확도 (Epoch)': train_acc_epoch,
                '테스트 정확도 (Trial)': test_acc_trial, '테스트 정확도 (Epoch)': test_acc_epoch,
                '학습 혼동 행렬 (Epoch)': cm_train_epoch.tolist() if cm_train_epoch is not None else None,
                '테스트 혼동 행렬 (Epoch)': cm_test_epoch.tolist() if cm_test_epoch is not None else None,
            })
            
# ======================================================================
# ✅ 최종 결과 출력 및 정렬 (U+00A0 제거 후 수정)
# ======================================================================

# 테스트 정확도 (Epoch) 기준 내림차순 정렬
all_results_sorted = sorted(all_results, key=lambda x: x['테스트 정확도 (Epoch)'], reverse=True)

# --------------------------------------------------
## 🥇 최종 분석 결과 요약 (상위 50개)
# --------------------------------------------------

print(f"\n================================================================")
print(f"      🥇 최종 자동화 분석 결과 요약 (상위 50개)")
print(f"      (테스트 정확도 (Epoch) 기준 내림차순 정렬)")
print(f"================================================================")

# 특징 이름 정리 함수 (밴드 추가)
def clean_feat_name(name):
    parts = name.split('_')
    # 단독 채널 (F_type_3d_Ch)
    if len(parts) == 4 and parts[-1] in CHANNELS:
        return f"{parts[1].capitalize()} {parts[-1]}"
    # 나머지
    return name.replace('F_', '').replace('_3d', '').replace('_', ' ').replace('area', '영역').replace('rm', 'ch').strip()

# 출력 테이블 포맷
table_data = []
for i, res in enumerate(all_results_sorted):
    if i >= 50: break
    table_data.append([
        f"{i+1}",
        f"{clean_feat_name(res['특징'])} ({res['밴드'].upper()})", # 특징 + 밴드
        res['정규화'],
        res['모델'],
        f"{res['학습 정확도 (Epoch)']:.4f}",
        f"{res['테스트 정확도 (Epoch)']:.4f}",
        f"{res['테스트 정확도 (Trial)']:.4f}",
    ])

# 테이블 출력 
headers = ["Rank", "특징 변수 (밴드)", "Norm", "모델", "Train Acc (Epoch)", "**Test Acc (Epoch)**", "Test Acc (Trial)"]
print(tabulate(table_data, headers=headers, tablefmt="fancy_grid", numalign="right"))


# --------------------------------------------------
## 🏆 최상위 3개 결과 상세 혼동 행렬
# --------------------------------------------------

print(f"\n================================================================")
print(f"      🏆 최상위 3개 결과 상세 혼동 행렬 (Epoch)")
print(f"================================================================")

for i in range(3):
    if i >= len(all_results_sorted): break
    res = all_results_sorted[i]
    
    # 특징/정규화/모델 이름 정리
    feat_band_clean = f"{clean_feat_name(res['특징'])} ({res['밴드'].upper()})"
    
    # 혼동 행렬 데이터 준비
    cm_train = np.array(res['학습 혼동 행렬 (Epoch)']) if res['학습 혼동 행렬 (Epoch)'] is not None else np.array([[0, 0], [0, 0]])
    cm_test = np.array(res['테스트 혼동 행렬 (Epoch)']) if res['테스트 혼동 행렬 (Epoch)'] is not None else np.array([[0, 0], [0, 0]])
    
    # 요청된 포맷에 맞춰 출력 (U+00A0 문자를 일반 공백으로 교체)
    print(f"--- Rank {i+1}. {feat_band_clean} | Norm: {res['정규화']} | Model: {res['모델']} ---")
    print(f"테스트 Epoch Acc: {res['테스트 정확도 (Epoch)']:.4f} / Trial Acc: {res['테스트 정확도 (Trial)']:.4f}")

    print(f"\n[Train CM (Epoch)] Acc: {res['학습 정확도 (Epoch)']:.4f}")
    print(f"True/Pred       True 0     True 1")
    print(f"Predicted 0           {cm_train[0, 0]:3.0f}     {cm_train[0, 1]:3.0f}")
    print(f"Predicted 1           {cm_train[1, 0]:3.0f}     {cm_train[1, 1]:3.0f}")

    print(f"\n[Test CM (Epoch)] Acc: {res['테스트 정확도 (Epoch)']:.4f}")
    print(f"True/Pred       True 0     True 1")
    print(f"Predicted 0           {cm_test[0, 0]:3.0f}     {cm_test[0, 1]:3.0f}")
    print(f"Predicted 1           {cm_test[1, 0]:3.0f}     {cm_test[1, 1]:3.0f}")
    if i < 2 and i < len(all_results_sorted) - 1:
        print() 

# --------------------------------------------------
## ⏱️ 실행 요약 (수정된 경우의 수 반영)
# --------------------------------------------------

time_taken_total = time.time() - start_time_total

print(f"\n================================================================")
print(f"                        ⏱️ 실행 요약")
print(f"================================================================")
print(f"총 결과 개수: {len(all_results)} ({total_cases}가지 경우)")
print(f"**전체 실행 시간**: {time_taken_total:.2f} 초")
print(f"평균 케이스당 시간: {time_taken_total / len(all_results):.4f} 초")
print(f"--------------------------------------------------")
print(f"분석 조건:")
print(f"  - 총 특징 쌍 개수: **86개** (단일 채널 48개 + 그룹 **38개**)")
print(f"  - 총 경우의 수: **9,396가지** (특징 86 * 밴드 6 * 정규화 3 * 모델 6)")
print(f"  - 훈련: 시각 데이터 전체 ({len(y)} epochs) - 정규화 적용됨")
print(f"  - 테스트: 후각 데이터 전체 ({len(y2)} epochs) - **정규화 적용됨**")
```
